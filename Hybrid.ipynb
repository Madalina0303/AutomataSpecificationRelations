```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
train_lambeq_pair_redundancy.py

End-to-end: CSV(sentence1,sentence2,label) -> lambeq QNLP pair classifier
- Reproducible shuffle
- 80/20 stratified split
- Tokenize (SpacyTokeniser) + parse (BobcatParser) -> diagrams
- UNK handling (UnknownWordsRewriteRule) + diagram simplification (RemoveCupsRewriter)
- IQPAnsatz -> quantum circuits
- Pair model (PennyLaneModel subclass): run 2 circuits, concat outputs, small NN head
- Train (PytorchTrainer), handle class imbalance (pos_weight OR oversampling)
- Evaluate (acc/prec/rec/F1) and export predictions + save model checkpoint

Tested design target: lambeq 0.5.x, Python 3.10+, CPU.
"""

import os
import json
import argparse
import random
from dataclasses import asdict, dataclass
from pathlib import Path
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import torch
from torch import nn

from tqdm import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    classification_report
)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from joblib import dump as joblib_dump

from lambeq import (
    BobcatParser,
    SpacyTokeniser,
    RemoveCupsRewriter,
    IQPAnsatz,
    AtomicType,
    UnknownWordsRewriteRule,
    Rewriter,
    Dataset,
    PennyLaneModel,
    PytorchTrainer,
)
from lambeq.backend.numerical_backend import set_backend as lambeq_set_backend


@dataclass
class Config:
    csv_path: str
    out_dir: str = "/content/outputs/run_pair_redundancy"
    seed: int = 42
    val_size: float = 0.20

    # Parsing / preprocessing
    min_freq_unk: int = 2
    parser_verbose: str = "suppress"  # 'suppress' or 'text'

    # Circuit / ansatz
    n_layers: int = 1
    n_single_qubit_params: int = 3

    # Training
    batch_size: int = 8
    epochs: int = 60
    learning_rate: float = 0.02
    early_stopping_patience: int = 10

    # Class imbalance handling
    imbalance_strategy: str = "pos_weight"  # 'pos_weight' or 'oversample' or 'none'

    # Baseline
    run_logreg_baseline: bool = True


def set_all_seeds(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def clean_text(x: object) -> Optional[str]:
    if not isinstance(x, str):
        return None
    s = " ".join(x.strip().split())
    return s if s else None


def load_and_shuffle_csv(csv_path: str, seed: int) -> pd.DataFrame:
    df = pd.read_csv(csv_path)

    required = {"sentence1", "sentence2", "label"}
    if not required.issubset(df.columns):
        raise ValueError(f"CSV must contain columns {sorted(required)}. Found: {list(df.columns)}")

    df = df.copy()
    df["sentence1"] = df["sentence1"].apply(clean_text)
    df["sentence2"] = df["sentence2"].apply(clean_text)

    df = df.dropna(subset=["sentence1", "sentence2", "label"]).reset_index(drop=True)

    # Validate labels
    df["label"] = df["label"].astype(int)
    if not set(df["label"].unique()).issubset({0, 1}):
        raise ValueError(f"Labels must be binary 0/1. Found: {sorted(df['label'].unique())}")

    # Reproducible shuffle
    df = df.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return df


def parse_sentence_list(
    sentences: List[str],
    parser: BobcatParser,
    tokeniser: SpacyTokeniser,
) -> Tuple[List[Optional[object]], List[Tuple[int, str]]]:
    diagrams: List[Optional[object]] = []
    failures: List[Tuple[int, str]] = []

    for i, s in enumerate(tqdm(sentences, desc="Parsing sentences", leave=False)):
        try:
            tokens = tokeniser.tokenise_sentence(s)
            diag = parser.sentence2diagram(tokens, tokenised=True)
            diagrams.append(diag)
        except Exception as e:
            diagrams.append(None)
            failures.append((i, repr(e)))

    return diagrams, failures


def oversample_train_df(train_df: pd.DataFrame, seed: int) -> pd.DataFrame:
    counts = train_df["label"].value_counts()
    if len(counts) < 2:
        return train_df

    maj_label = int(counts.idxmax())
    min_label = int(counts.idxmin())
    n_maj = int(counts[maj_label])
    n_min = int(counts[min_label])

    if n_min == n_maj:
        return train_df

    n_to_add = n_maj - n_min
    min_df = train_df[train_df["label"] == min_label]
    add_df = min_df.sample(n=n_to_add, replace=True, random_state=seed)

    out = pd.concat([train_df, add_df], ignore_index=True)
    out = out.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return out


class PairRedundancyModel(PennyLaneModel):
    """Hybrid model: two circuits -> concat probs -> small NN -> logit"""

    def __init__(self, hidden_dim: int = 16, dropout: float = 0.10, **kwargs):
        super().__init__(**kwargs)
        self.net = nn.Sequential(
            nn.Linear(4, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 1)  # logit
        )

    def forward(self, diagram_pairs: List[Tuple[object, object]]) -> torch.Tensor:
        # diagram_pairs is a list of tuples (circ1, circ2)
        first, second = zip(*diagram_pairs)
        out1 = self.get_diagram_output(list(first))   # shape [B,2] for 1-qubit prob
        out2 = self.get_diagram_output(list(second))  # shape [B,2]
        x = torch.cat([out1, out2], dim=1)            # shape [B,4]

        # map [0,1] -> [-1,1] (helps ReLU)
        x = 2.0 * (x - 0.5)

        logits = self.net(x).squeeze(-1)              # shape [B]
        return logits


def evaluate_binary(y_true: np.ndarray, prob_pos: np.ndarray) -> dict:
    y_pred = (prob_pos >= 0.5).astype(int)

    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, _support = precision_recall_fscore_support(
        y_true, y_pred, average="binary", pos_label=1, zero_division=0
    )
    cm = confusion_matrix(y_true, y_pred)

    return {
        "accuracy": float(acc),
        "precision_pos1": float(prec),
        "recall_pos1": float(rec),
        "f1_pos1": float(f1),
        "confusion_matrix": cm.tolist(),
        "classification_report": classification_report(
            y_true, y_pred, digits=4, zero_division=0
        ),
    }


def main(cfg: Config) -> None:
    os.environ["TOKENIZERS_PARALLELISM"] = "true"
    set_all_seeds(cfg.seed)

    out_dir = Path(cfg.out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    # Save config for reproducibility
    with (out_dir / "config.json").open("w", encoding="utf-8") as f:
        json.dump(asdict(cfg), f, ensure_ascii=False, indent=2)

    # Load and split
    df = load_and_shuffle_csv(cfg.csv_path, cfg.seed)
    train_df, val_df = train_test_split(
        df,
        test_size=cfg.val_size,
        random_state=cfg.seed,
        stratify=df["label"],
        shuffle=True,
    )
    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)

    # Optional: oversample to handle imbalance
    if cfg.imbalance_strategy == "oversample":
        train_df = oversample_train_df(train_df, cfg.seed)

    # Tokeniser + parser
    tokeniser = SpacyTokeniser()
    parser = BobcatParser(verbose=cfg.parser_verbose)

    # Parse all sentences (train + val) for sentence1 and sentence2
    all_s1 = pd.concat([train_df["sentence1"], val_df["sentence1"]], ignore_index=True).tolist()
    all_s2 = pd.concat([train_df["sentence2"], val_df["sentence2"]], ignore_index=True).tolist()

    diags1, fail1 = parse_sentence_list(all_s1, parser, tokeniser)
    diags2, fail2 = parse_sentence_list(all_s2, parser, tokeniser)

    # Build a combined dataframe back
    all_pairs_df = pd.concat([train_df, val_df], ignore_index=True).copy()
    all_pairs_df["diag1"] = diags1
    all_pairs_df["diag2"] = diags2

    # Drop rows with parsing failures
    before = len(all_pairs_df)
    all_pairs_df = all_pairs_df.dropna(subset=["diag1", "diag2"]).reset_index(drop=True)
    dropped = before - len(all_pairs_df)

    with (out_dir / "parsing_stats.json").open("w", encoding="utf-8") as f:
        json.dump(
            {
                "initial_rows": int(before),
                "kept_rows": int(len(all_pairs_df)),
                "dropped_rows_due_to_parsing": int(dropped),
                "fail_sentence1_count": int(len(fail1)),
                "fail_sentence2_count": int(len(fail2)),
                "fail_sentence1_examples": fail1[:10],
                "fail_sentence2_examples": fail2[:10],
            },
            f, ensure_ascii=False, indent=2
        )
uit(d) for d in tqdm(split_df["diag2"], desc=f"Circuits {split_name} s2", leave=False)]

    # Prepare datasets (lambeq Dataset)
    train_pairs = list(zip(train_df2["circ1"].tolist(), train_df2["circ2"].tolist()))
    val_pairs = list(zip(val_df2["circ1"].tolist(), val_df2["circ2"].tolist()))

    y_train = train_df2["label"].astype(float).tolist()
    y_val = val_df2["label"].astype(float).tolist()

    # Set lambeq numerical backend to PyTorch so Dataset yields torch tensors
    lambeq_set_backend("pytorch")

    train_dataset = Dataset(train_pairs, y_train, batch_size=cfg.batch_size, shuffle=True)
    val_dataset = Dataset(val_pairs, y_val, batch_size=cfg.batch_size, shuffle=False)

    # Build model from ALL circuits we will use (train+val) so circuit_map is complete
    all_circuits = list(train_df2["circ1"]) + list(train_df2["circ2"]) + list(val_df2["circ1"]) + list(val_df2["circ2"])

    backend_config = {"backend": "default.qubit"}  # local simulator

    model = PairRedundancyModel.from_diagrams(
        all_circuits,
        probabilities=True,
        normalize=True,
        backend_config=backend_config,
        hidden_dim=16,
        dropout=0.10,
    )
    model.initialise_weights()

    # Class imbalance handling
    n_pos = int(sum(y_train))
    n_neg = int(len(y_train) - n_pos)
    pos_weight_value = (n_neg / max(n_pos, 1)) if cfg.imbalance_strategy == "pos_weight" else 1.0
    pos_weight = torch.tensor([pos_weight_value], dt
    # Re-split (after dropping unparsable rows) with same val_size, stratified
    train_df2, val_df2 = train_test_split(
        all_pairs_df,
        test_size=cfg.val_size,
        random_state=cfg.seed,
        stratify=all_pairs_df["label"],
        shuffle=True,
    )
    train_df2 = train_df2.reset_index(drop=True)
    val_df2 = val_df2.reset_index(drop=True)

    # Unknown words rule from TRAIN diagrams only
    train_diagrams_for_vocab = (
        train_df2["diag1"].tolist() + train_df2["diag2"].tolist()
    )
    unk_rule = UnknownWordsRewriteRule.from_diagrams(
        diagrams=train_diagrams_for_vocab,
        min_freq=cfg.min_freq_unk,
        ignore_types=True
    )
    rewriter = Rewriter([unk_rule])

    # Save vocabulary for later inference
    vocab = sorted(list(unk_rule.vocabulary))
    with (out_dir / "unk_vocabulary.json").open("w", encoding="utf-8") as f:
        json.dump(
            {"min_freq": cfg.min_freq_unk, "ignore_types": True, "vocabulary": vocab},
            f, ensure_ascii=False, indent=2
        )

    # Simplify diagrams + build circuits
    remove_cups = RemoveCupsRewriter()
    ansatz = IQPAnsatz(
        {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1, AtomicType.PREPOSITIONAL_PHRASE:0},
        n_layers=cfg.n_layers,
        n_single_qubit_params=cfg.n_single_qubit_params
    )

    def diag_to_circuit(diag_obj: object) -> object:
        d = rewriter(diag_obj)
        d = remove_cups(d)
        return ansatz(d)

    for split_name, split_df in [("train", train_df2), ("val", val_df2)]:
        split_df["circ1"] = [diag_to_circuit(d) for d in tqdm(split_df["diag1"], desc=f"Circuits {split_name} s1", leave=False)]
        split_df["circ2"] = [diag_to_circype=torch.float32)

    def loss_fn(logits: torch.Tensor, y: torch.Tensor) -> torch.Tensor:
        y = y.float().view(-1)
        return torch.nn.functional.binary_cross_entropy_with_logits(
            logits, y, pos_weight=pos_weight.to(logits.device)
        )

    def acc_fn(logits: torch.Tensor, y: torch.Tensor) -> float:
        y = y.float().view(-1)
        preds = (torch.sigmoid(logits) >= 0.5).float()
        return float((preds == y).float().mean().item())

    trainer = PytorchTrainer(
        model=model,
        loss_function=loss_fn,
        optimizer=torch.optim.Adam,
        learning_rate=cfg.learning_rate,
        epochs=cfg.epochs,
        evaluate_functions={"acc": acc_fn},
        evaluate_on_train=True,
        use_tensorboard=False,
        log_dir=str(out_dir),
        verbose="text",
        seed=cfg.seed
    )

    # Train with early stopping on validation accuracy
    trainer.fit(
        train_dataset,
        val_dataset,
        early_stopping_criterion="acc",
        early_stopping_interval=cfg.early_stopping_patience,
        minimize_criterion=False
    )

    # Save final checkpoint explicitly (even if best_model exists)
    final_ckpt = out_dir / "final_model.lt"
    model.save(final_ckpt.as_posix())

    # Prefer best_model.lt if it exists
    best_ckpt = out_dir / "best_model.lt"
    ckpt_to_use = best_ckpt if best_ckpt.exists() else final_ckpt

    # Load chosen checkpoint into the same model instance
    model.load(ckpt_to_use.as_posix())

    # Evaluate on validation set
    model.eval()
    with torch.no_grad():
        val_logits = model(val_pairs).detach().cpu().numpy()
        val_prob = 1.0 / (1.0 + np.exp(-val_logits))

    y_val_np = np.array(y_val, dtype=int)
    metrics = evaluate_binary(y_val_np, val_prob)

    with (out_dir / "metrics_val.json").open("w", encoding="utf-8") as f:
        json.dump(metrics, f, ensure_ascii=False, indent=2)

    print("\n=== Validation metrics (QNLP pair model) ===")
    print(metrics["classification_report"])
    print("Confusion matrix:\n", np.array(metrics["confusion_matrix"]))
    print("Saved checkpoint used for eval:", ckpt_to_use.as_posix())

    # Export predictions for train+val
    full_df = pd.concat([train_df2.assign(split="train"), val_df2.assign(split="val")], ignore_index=True)
    full_pairs = list(zip(full_df["circ1"].tolist(), full_df["circ2"].tolist()))
    y_true_full = full_df["label"].astype(int).to_numpy()

    with torch.no_grad():
        logits_full = model(full_pairs).detach().cpu().numpy()
        prob_full = 1.0 / (1.0 + np.exp(-logits_full))
        pred_full = (prob_full >= 0.5).astype(int)

    pred_out = pd.DataFrame({
        "sentence1": full_df["sentence1"].tolist(),
        "sentence2": full_df["sentence2"].tolist(),
        "label_true": y_true_full,
        "split": full_df["split"].tolist(),
        "prob_redundancy": prob_full,
        "label_pred": pred_full,
    })
    pred_out.to_csv(out_dir / "predictions_qnlp.csv", index=False)

    # Optional baseline: TF-IDF + Logistic Regression
    if cfg.run_logreg_baseline:
        train_texts = (train_df2["sentence1"] + " [SEP] " + train_df2["sentence2"]).tolist()
        val_texts = (val_df2["sentence1"] + " [SEP] " + val_df2["sentence2"]).tolist()
        y_train_lr = train_df2["label"].astype(int).to_numpy()
        y_val_lr = val_df2["label"].astype(int).to_numpy()

        vectorizer = TfidfVectorizer(ngram_range=(1, 2), max_features=5000)
        X_train = vectorizer.fit_transform(train_texts)
        X_val = vectorizer.transform(val_texts)

        clf = LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=cfg.seed
        )
        clf.fit(X_train, y_train_lr)

        val_pred = clf.predict(X_val)
        val_prob_lr = clf.predict_proba(X_val)[:, 1]

        baseline_metrics = evaluate_binary(y_val_lr, val_prob_lr)
        with (out_dir / "metrics_val_logreg.json").open("w", encoding="utf-8") as f:
            json.dump(baseline_metrics, f, ensure_ascii=False, indent=2)

        joblib_dump(vectorizer, out_dir / "tfidf_vectorizer.joblib")
        joblib_dump(clf, out_dir / "logreg_model.joblib")

        print("\n=== Validation metrics (LogReg baseline) ===")
        print(baseline_metrics["classification_report"])

    print("\nArtifacts written to:", out_dir.resolve())


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--csv_path", type=str, default="/content/redudancy.csv")
    parser.add_argument("--out_dir", type=str, default="outputs/run_pair_redundancy")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--val_size", type=float, default=0.20)
    parser.add_argument("--min_freq_unk", type=int, default=2)
    parser.add_argument("--batch_size", type=int, default=8)
    parser.add_argument("--epochs", type=int, default=60)
    parser.add_argument("--learning_rate", type=float, default=0.02)
    parser.add_argument("--early_stopping_patience", type=int, default=10)
    parser.add_argument("--imbalance_strategy", type=str, default="pos_weight",
                        choices=["pos_weight", "oversample", "none"])
    parser.add_argument("--run_logreg_baseline", action="store_true")

    # 🔥 FIX pentru Jupyter / Colab
    args, _ = parser.parse_known_args()

    cfg = Config(**vars(args))
    main(cfg)
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
train_lambeq_pair_redundancy.py

End-to-end: CSV(sentence1,sentence2,label) -> lambeq QNLP pair classifier
- Reproducible shuffle
- 80/20 stratified split
- Tokenize (SpacyTokeniser) + parse (BobcatParser) -> diagrams
- UNK handling (UnknownWordsRewriteRule) + diagram simplification (RemoveCupsRewriter)
- IQPAnsatz -> quantum circuits
- Pair model (PennyLaneModel subclass): run 2 circuits, concat outputs, small NN head
- Train (PytorchTrainer), handle class imbalance (pos_weight OR oversampling)
- Evaluate (acc/prec/rec/F1) and export predictions + save model checkpoint

Tested design target: lambeq 0.5.x, Python 3.10+, CPU.
"""

import os
import json
import argparse
import random
from dataclasses import asdict, dataclass
from pathlib import Path
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import torch
from torch import nn

from tqdm import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    classification_report
)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from joblib import dump as joblib_dump

from lambeq import (
    BobcatParser,
    SpacyTokeniser,
    RemoveCupsRewriter,
    IQPAnsatz,
    AtomicType,
    UnknownWordsRewriteRule,
    Rewriter,
    Dataset,
    PennyLaneModel,
    PytorchTrainer,
)
from lambeq.backend.numerical_backend import set_backend as lambeq_set_backend


@dataclass
class Config:
    csv_path: str
    out_dir: str = "/content/outputs/run_pair_contr"
    seed: int = 42
    val_size: float = 0.20

    # Parsing / preprocessing
    min_freq_unk: int = 2
    parser_verbose: str = "suppress"  # 'suppress' or 'text'

    # Circuit / ansatz
    n_layers: int = 1
    n_single_qubit_params: int = 3

    # Training
    batch_size: int = 8
    epochs: int = 60
    learning_rate: float = 0.02
    early_stopping_patience: int = 10

    # Class imbalance handling
    imbalance_strategy: str = "pos_weight"  # 'pos_weight' or 'oversample' or 'none'

    # Baseline
    run_logreg_baseline: bool = True


def set_all_seeds(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def clean_text(x: object) -> Optional[str]:
    if not isinstance(x, str):
        return None
    s = " ".join(x.strip().split())
    return s if s else None


def load_and_shuffle_csv(csv_path: str, seed: int) -> pd.DataFrame:
    df = pd.read_csv(csv_path)

    required = {"sentence1", "sentence2", "label"}
    if not required.issubset(df.columns):
        raise ValueError(f"CSV must contain columns {sorted(required)}. Found: {list(df.columns)}")

    df = df.copy()
    df["sentence1"] = df["sentence1"].apply(clean_text)
    df["sentence2"] = df["sentence2"].apply(clean_text)

    df = df.dropna(subset=["sentence1", "sentence2", "label"]).reset_index(drop=True)

    # Validate labels
    df["label"] = df["label"].astype(int)
    if not set(df["label"].unique()).issubset({0, 1}):
        raise ValueError(f"Labels must be binary 0/1. Found: {sorted(df['label'].unique())}")

    # Reproducible shuffle
    df = df.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return df


def parse_sentence_list(
    sentences: List[str],
    parser: BobcatParser,
    tokeniser: SpacyTokeniser,
) -> Tuple[List[Optional[object]], List[Tuple[int, str]]]:
    diagrams: List[Optional[object]] = []
    failures: List[Tuple[int, str]] = []

    for i, s in enumerate(tqdm(sentences, desc="Parsing sentences", leave=False)):
        try:
            tokens = tokeniser.tokenise_sentence(s)
            diag = parser.sentence2diagram(tokens, tokenised=True)
            diagrams.append(diag)
        except Exception as e:
            diagrams.append(None)
            failures.append((i, repr(e)))

    return diagrams, failures


def oversample_train_df(train_df: pd.DataFrame, seed: int) -> pd.DataFrame:
    counts = train_df["label"].value_counts()
    if len(counts) < 2:
        return train_df

    maj_label = int(counts.idxmax())
    min_label = int(counts.idxmin())
    n_maj = int(counts[maj_label])
    n_min = int(counts[min_label])

    if n_min == n_maj:
        return train_df

    n_to_add = n_maj - n_min
    min_df = train_df[train_df["label"] == min_label]
    add_df = min_df.sample(n=n_to_add, replace=True, random_state=seed)

    out = pd.concat([train_df, add_df], ignore_index=True)
    out = out.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return out


class PairRedundancyModel(PennyLaneModel):
    """Hybrid model: two circuits -> concat probs -> small NN -> logit"""

    def __init__(self, hidden_dim: int = 16, dropout: float = 0.10, **kwargs):
        super().__init__(**kwargs)
        self.net = nn.Sequential(
            nn.Linear(4, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 1)  # logit
        )

    def forward(self, diagram_pairs: List[Tuple[object, object]]) -> torch.Tensor:
        # diagram_pairs is a list of tuples (circ1, circ2)
        first, second = zip(*diagram_pairs)
        out1 = self.get_diagram_output(list(first))   # shape [B,2] for 1-qubit prob
        out2 = self.get_diagram_output(list(second))  # shape [B,2]
        x = torch.cat([out1, out2], dim=1)            # shape [B,4]

        # map [0,1] -> [-1,1] (helps ReLU)
        x = 2.0 * (x - 0.5)

        logits = self.net(x).squeeze(-1)              # shape [B]
        return logits


def evaluate_binary(y_true: np.ndarray, prob_pos: np.ndarray) -> dict:
    y_pred = (prob_pos >= 0.5).astype(int)

    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, _support = precision_recall_fscore_support(
        y_true, y_pred, average="binary", pos_label=1, zero_division=0
    )
    cm = confusion_matrix(y_true, y_pred)

    return {
        "accuracy": float(acc),
        "precision_pos1": float(prec),
        "recall_pos1": float(rec),
        "f1_pos1": float(f1),
        "confusion_matrix": cm.tolist(),
        "classification_report": classification_report(
            y_true, y_pred, digits=4, zero_division=0
        ),
    }


def main(cfg: Config) -> None:
    os.environ["TOKENIZERS_PARALLELISM"] = "true"
    set_all_seeds(cfg.seed)

    out_dir = Path(cfg.out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    # Save config for reproducibility
    with (out_dir / "config.json").open("w", encoding="utf-8") as f:
        json.dump(asdict(cfg), f, ensure_ascii=False, indent=2)

    # Load and split
    df = load_and_shuffle_csv(cfg.csv_path, cfg.seed)
    train_df, val_df = train_test_split(
        df,
        test_size=cfg.val_size,
        random_state=cfg.seed,
        stratify=df["label"],
        shuffle=True,
    )
    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)

    # Optional: oversample to handle imbalance
    if cfg.imbalance_strategy == "oversample":
        train_df = oversample_train_df(train_df, cfg.seed)

    # Tokeniser + parser
    tokeniser = SpacyTokeniser()
    parser = BobcatParser(verbose=cfg.parser_verbose)

    # Parse all sentences (train + val) for sentence1 and sentence2
    all_s1 = pd.concat([train_df["sentence1"], val_df["sentence1"]], ignore_index=True).tolist()
    all_s2 = pd.concat([train_df["sentence2"], val_df["sentence2"]], ignore_index=True).tolist()

    diags1, fail1 = parse_sentence_list(all_s1, parser, tokeniser)
    diags2, fail2 = parse_sentence_list(all_s2, parser, tokeniser)

    # Build a combined dataframe back
    all_pairs_df = pd.concat([train_df, val_df], ignore_index=True).copy()
    all_pairs_df["diag1"] = diags1
    all_pairs_df["diag2"] = diags2

    # Drop rows with parsing failures
    before = len(all_pairs_df)
    all_pairs_df = all_pairs_df.dropna(subset=["diag1", "diag2"]).reset_index(drop=True)
    dropped = before - len(all_pairs_df)

    with (out_dir / "parsing_stats.json").open("w", encoding="utf-8") as f:
        json.dump(
            {
                "initial_rows": int(before),
                "kept_rows": int(len(all_pairs_df)),
                "dropped_rows_due_to_parsing": int(dropped),
                "fail_sentence1_count": int(len(fail1)),
                "fail_sentence2_count": int(len(fail2)),
                "fail_sentence1_examples": fail1[:10],
                "fail_sentence2_examples": fail2[:10],
            },
            f, ensure_ascii=False, indent=2
        )

    # Re-split (after dropping unparsable rows) with same val_size, stratified
    train_df2, val_df2 = train_test_split(
        all_pairs_df,
        test_size=cfg.val_size,
        random_state=cfg.seed,
        stratify=all_pairs_df["label"],
        shuffle=True,
    )
    train_df2 = train_df2.reset_index(drop=True)
    val_df2 = val_df2.reset_index(drop=True)

    # Unknown words rule from TRAIN diagrams only
    train_diagrams_for_vocab = (
        train_df2["diag1"].tolist() + train_df2["diag2"].tolist()
    )
    unk_rule = UnknownWordsRewriteRule.from_diagrams(
        diagrams=train_diagrams_for_vocab,
        min_freq=cfg.min_freq_unk,
        ignore_types=True
    )
    rewriter = Rewriter([unk_rule])

    # Save vocabulary for later inference
    vocab = sorted(list(unk_rule.vocabulary))
    with (out_dir / "unk_vocabulary.json").open("w", encoding="utf-8") as f:
        json.dump(
            {"min_freq": cfg.min_freq_unk, "ignore_types": True, "vocabulary": vocab},
            f, ensure_ascii=False, indent=2
        )

    # Simplify diagrams + build circuits
    remove_cups = RemoveCupsRewriter()
    ansatz = IQPAnsatz(
        {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1, AtomicType.PREPOSITIONAL_PHRASE:0},
        n_layers=cfg.n_layers,
        n_single_qubit_params=cfg.n_single_qubit_params
    )

    def diag_to_circuit(diag_obj: object) -> object:
        d = rewriter(diag_obj)
        d = remove_cups(d)
        return ansatz(d)

    for split_name, split_df in [("train", train_df2), ("val", val_df2)]:
        split_df["circ1"] = [diag_to_circuit(d) for d in tqdm(split_df["diag1"], desc=f"Circuits {split_name} s1", leave=False)]
        split_df["circ2"] = [diag_to_circuit(d) for d in tqdm(split_df["diag2"], desc=f"Circuits {split_name} s2", leave=False)]

    # Prepare datasets (lambeq Dataset)
    train_pairs = list(zip(train_df2["circ1"].tolist(), train_df2["circ2"].tolist()))
    val_pairs = list(zip(val_df2["circ1"].tolist(), val_df2["circ2"].tolist()))

    y_train = train_df2["label"].astype(float).tolist()
    y_val = val_df2["label"].astype(float).tolist()

    # Set lambeq numerical backend to PyTorch so Dataset yields torch tensors
    lambeq_set_backend("pytorch")

    train_dataset = Dataset(train_pairs, y_train, batch_size=cfg.batch_size, shuffle=True)
    val_dataset = Dataset(val_pairs, y_val, batch_size=cfg.batch_size, shuffle=False)

    # Build model from ALL circuits we will use (train+val) so circuit_map is complete
    all_circuits = list(train_df2["circ1"]) + list(train_df2["circ2"]) + list(val_df2["circ1"]) + list(val_df2["circ2"])

    backend_config = {"backend": "default.qubit"}  # local simulator

    model = PairRedundancyModel.from_diagrams(
        all_circuits,
        probabilities=True,
        normalize=True,
        backend_config=backend_config,
        hidden_dim=16,
        dropout=0.10,
    )
    model.initialise_weights()

    # Class imbalance handling
    n_pos = int(sum(y_train))
    n_neg = int(len(y_train) - n_pos)
    pos_weight_value = (n_neg / max(n_pos, 1)) if cfg.imbalance_strategy == "pos_weight" else 1.0
    pos_weight = torch.tensor([pos_weight_value], dtype=torch.float32)

    def loss_fn(logits: torch.Tensor, y: torch.Tensor) -> torch.Tensor:
        y = y.float().view(-1)
        return torch.nn.functional.binary_cross_entropy_with_logits(
            logits, y, pos_weight=pos_weight.to(logits.device)
        )

    def acc_fn(logits: torch.Tensor, y: torch.Tensor) -> float:
        y = y.float().view(-1)
        preds = (torch.sigmoid(logits) >= 0.5).float()
        return float((preds == y).float().mean().item())

    trainer = PytorchTrainer(
        model=model,
        loss_function=loss_fn,
        optimizer=torch.optim.Adam,
        learning_rate=cfg.learning_rate,
        epochs=cfg.epochs,
        evaluate_functions={"acc": acc_fn},
        evaluate_on_train=True,
        use_tensorboard=False,
        log_dir=str(out_dir),
        verbose="text",
        seed=cfg.seed
    )

    # Train with early stopping on validation accuracy
    trainer.fit(
        train_dataset,
        val_dataset,
        early_stopping_criterion="acc",
        early_stopping_interval=cfg.early_stopping_patience,
        minimize_criterion=False
    )

    # Save final checkpoint explicitly (even if best_model exists)
    final_ckpt = out_dir / "final_model.lt"
    model.save(final_ckpt.as_posix())

    # Prefer best_model.lt if it exists
    best_ckpt = out_dir / "best_model.lt"
    ckpt_to_use = best_ckpt if best_ckpt.exists() else final_ckpt

    # Load chosen checkpoint into the same model instance
    model.load(ckpt_to_use.as_posix())

    # Evaluate on validation set
    model.eval()
    with torch.no_grad():
        val_logits = model(val_pairs).detach().cpu().numpy()
        val_prob = 1.0 / (1.0 + np.exp(-val_logits))

    y_val_np = np.array(y_val, dtype=int)
    metrics = evaluate_binary(y_val_np, val_prob)

    with (out_dir / "metrics_val.json").open("w", encoding="utf-8") as f:
        json.dump(metrics, f, ensure_ascii=False, indent=2)

    print("\n=== Validation metrics (QNLP pair model) ===")
    print(metrics["classification_report"])
    print("Confusion matrix:\n", np.array(metrics["confusion_matrix"]))
    print("Saved checkpoint used for eval:", ckpt_to_use.as_posix())

    # Export predictions for train+val
    full_df = pd.concat([train_df2.assign(split="train"), val_df2.assign(split="val")], ignore_index=True)
    full_pairs = list(zip(full_df["circ1"].tolist(), full_df["circ2"].tolist()))
    y_true_full = full_df["label"].astype(int).to_numpy()

    with torch.no_grad():
        logits_full = model(full_pairs).detach().cpu().numpy()
        prob_full = 1.0 / (1.0 + np.exp(-logits_full))
        pred_full = (prob_full >= 0.5).astype(int)

    pred_out = pd.DataFrame({
        "sentence1": full_df["sentence1"].tolist(),
        "sentence2": full_df["sentence2"].tolist(),
        "label_true": y_true_full,
        "split": full_df["split"].tolist(),
        "prob_redundancy": prob_full,
        "label_pred": pred_full,
    })
    pred_out.to_csv(out_dir / "predictions_qnlp.csv", index=False)

    # Optional baseline: TF-IDF + Logistic Regression
    if cfg.run_logreg_baseline:
        train_texts = (train_df2["sentence1"] + " [SEP] " + train_df2["sentence2"]).tolist()
        val_texts = (val_df2["sentence1"] + " [SEP] " + val_df2["sentence2"]).tolist()
        y_train_lr = train_df2["label"].astype(int).to_numpy()
        y_val_lr = val_df2["label"].astype(int).to_numpy()

        vectorizer = TfidfVectorizer(ngram_range=(1, 2), max_features=5000)
        X_train = vectorizer.fit_transform(train_texts)
        X_val = vectorizer.transform(val_texts)

        clf = LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=cfg.seed
        )
        clf.fit(X_train, y_train_lr)

        val_pred = clf.predict(X_val)
        val_prob_lr = clf.predict_proba(X_val)[:, 1]

        baseline_metrics = evaluate_binary(y_val_lr, val_prob_lr)
        with (out_dir / "metrics_val_logreg.json").open("w", encoding="utf-8") as f:
            json.dump(baseline_metrics, f, ensure_ascii=False, indent=2)

        joblib_dump(vectorizer, out_dir / "tfidf_vectorizer.joblib")
        joblib_dump(clf, out_dir / "logreg_model.joblib")

        print("\n=== Validation metrics (LogReg baseline) ===")
        print(baseline_metrics["classification_report"])

    print("\nArtifacts written to:", out_dir.resolve())


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--csv_path", type=str, default="/content/NU.csv")
    parser.add_argument("--out_dir", type=str, default="/content/outputs/run_pair_contradiction")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--val_size", type=float, default=0.20)
    parser.add_argument("--min_freq_unk", type=int, default=2)
    parser.add_argument("--batch_size", type=int, default=8)
    parser.add_argument("--epochs", type=int, default=60)
    parser.add_argument("--learning_rate", type=float, default=0.02)
    parser.add_argument("--early_stopping_patience", type=int, default=10)
    parser.add_argument("--imbalance_strategy", type=str, default="pos_weight",
                        choices=["pos_weight", "oversample", "none"])
    parser.add_argument("--run_logreg_baseline", action="store_true")

    # 🔥 FIX pentru Jupyter / Colab
    args, _ = parser.parse_known_args()

    cfg = Config(**vars(args))
    main(cfg)
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
train_lambeq_pair_redundancy.py

End-to-end: CSV(sentence1,sentence2,label) -> lambeq QNLP pair classifier
- Reproducible shuffle
- 80/20 stratified split
- Tokenize (SpacyTokeniser) + parse (BobcatParser) -> diagrams
- UNK handling (UnknownWordsRewriteRule) + diagram simplification (RemoveCupsRewriter)
- IQPAnsatz -> quantum circuits
- Pair model (PennyLaneModel subclass): run 2 circuits, concat outputs, small NN head
- Train (PytorchTrainer), handle class imbalance (pos_weight OR oversampling)
- Evaluate (acc/prec/rec/F1) and export predictions + save model checkpoint

Tested design target: lambeq 0.5.x, Python 3.10+, CPU.
"""

import os
import json
import argparse
import random
from dataclasses import asdict, dataclass
from pathlib import Path
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import torch
from torch import nn

from tqdm import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    classification_report
)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from joblib import dump as joblib_dump

from lambeq import (
    BobcatParser,
    SpacyTokeniser,
    RemoveCupsRewriter,
    IQPAnsatz,
    AtomicType,
    UnknownWordsRewriteRule,
    Rewriter,
    Dataset,
    PennyLaneModel,
    PytorchTrainer,
)
from lambeq.backend.numerical_backend import set_backend as lambeq_set_backend


@dataclass
class Config:
    csv_path: str
    out_dir: str = "/content/outputs/run_pair_exclusion"
    seed: int = 42
    val_size: float = 0.20

    # Parsing / preprocessing
    min_freq_unk: int = 2
    parser_verbose: str = "suppress"  # 'suppress' or 'text'

    # Circuit / ansatz
    n_layers: int = 1
    n_single_qubit_params: int = 3

    # Training
    batch_size: int = 8
    epochs: int = 60
    learning_rate: float = 0.02
    early_stopping_patience: int = 10

    # Class imbalance handling
    imbalance_strategy: str = "pos_weight"  # 'pos_weight' or 'oversample' or 'none'

    # Baseline
    run_logreg_baseline: bool = True


def set_all_seeds(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def clean_text(x: object) -> Optional[str]:
    if not isinstance(x, str):
        return None
    s = " ".join(x.strip().split())
    return s if s else None


def load_and_shuffle_csv(csv_path: str, seed: int) -> pd.DataFrame:
    df = pd.read_csv(csv_path)

    required = {"sentence1", "sentence2", "label"}
    if not required.issubset(df.columns):
        raise ValueError(f"CSV must contain columns {sorted(required)}. Found: {list(df.columns)}")

    df = df.copy()
    df["sentence1"] = df["sentence1"].apply(clean_text)
    df["sentence2"] = df["sentence2"].apply(clean_text)

    df = df.dropna(subset=["sentence1", "sentence2", "label"]).reset_index(drop=True)

    # Validate labels
    df["label"] = df["label"].astype(int)
    if not set(df["label"].unique()).issubset({0, 1}):
        raise ValueError(f"Labels must be binary 0/1. Found: {sorted(df['label'].unique())}")

    # Reproducible shuffle
    df = df.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return df


def parse_sentence_list(
    sentences: List[str],
    parser: BobcatParser,
    tokeniser: SpacyTokeniser,
) -> Tuple[List[Optional[object]], List[Tuple[int, str]]]:
    diagrams: List[Optional[object]] = []
    failures: List[Tuple[int, str]] = []

    for i, s in enumerate(tqdm(sentences, desc="Parsing sentences", leave=False)):
        try:
            tokens = tokeniser.tokenise_sentence(s)
            diag = parser.sentence2diagram(tokens, tokenised=True)
            diagrams.append(diag)
        except Exception as e:
            diagrams.append(None)
            failures.append((i, repr(e)))

    return diagrams, failures


def oversample_train_df(train_df: pd.DataFrame, seed: int) -> pd.DataFrame:
    counts = train_df["label"].value_counts()
    if len(counts) < 2:
        return train_df

    maj_label = int(counts.idxmax())
    min_label = int(counts.idxmin())
    n_maj = int(counts[maj_label])
    n_min = int(counts[min_label])

    if n_min == n_maj:
        return train_df

    n_to_add = n_maj - n_min
    min_df = train_df[train_df["label"] == min_label]
    add_df = min_df.sample(n=n_to_add, replace=True, random_state=seed)

    out = pd.concat([train_df, add_df], ignore_index=True)
    out = out.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return out


class PairRedundancyModel(PennyLaneModel):
    """Hybrid model: two circuits -> concat probs -> small NN -> logit"""

    def __init__(self, hidden_dim: int = 16, dropout: float = 0.10, **kwargs):
        super().__init__(**kwargs)
        self.net = nn.Sequential(
            nn.Linear(4, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 1)  # logit
        )

    def forward(self, diagram_pairs: List[Tuple[object, object]]) -> torch.Tensor:
        # diagram_pairs is a list of tuples (circ1, circ2)
        first, second = zip(*diagram_pairs)
        out1 = self.get_diagram_output(list(first))   # shape [B,2] for 1-qubit prob
        out2 = self.get_diagram_output(list(second))  # shape [B,2]
        x = torch.cat([out1, out2], dim=1)            # shape [B,4]

        # map [0,1] -> [-1,1] (helps ReLU)
        x = 2.0 * (x - 0.5)

        logits = self.net(x).squeeze(-1)              # shape [B]
        return logits


def evaluate_binary(y_true: np.ndarray, prob_pos: np.ndarray) -> dict:
    y_pred = (prob_pos >= 0.5).astype(int)

    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, _support = precision_recall_fscore_support(
        y_true, y_pred, average="binary", pos_label=1, zero_division=0
    )
    cm = confusion_matrix(y_true, y_pred)

    return {
        "accuracy": float(acc),
        "precision_pos1": float(prec),
        "recall_pos1": float(rec),
        "f1_pos1": float(f1),
        "confusion_matrix": cm.tolist(),
        "classification_report": classification_report(
            y_true, y_pred, digits=4, zero_division=0
        ),
    }


def main(cfg: Config) -> None:
    os.environ["TOKENIZERS_PARALLELISM"] = "true"
    set_all_seeds(cfg.seed)

    out_dir = Path(cfg.out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    # Save config for reproducibility
    with (out_dir / "config.json").open("w", encoding="utf-8") as f:
        json.dump(asdict(cfg), f, ensure_ascii=False, indent=2)

    # Load and split
    df = load_and_shuffle_csv(cfg.csv_path, cfg.seed)
    train_df, val_df = train_test_split(
        df,
        test_size=cfg.val_size,
        random_state=cfg.seed,
        stratify=df["label"],
        shuffle=True,
    )
    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)

    # Optional: oversample to handle imbalance
    if cfg.imbalance_strategy == "oversample":
        train_df = oversample_train_df(train_df, cfg.seed)

    # Tokeniser + parser
    tokeniser = SpacyTokeniser()
    parser = BobcatParser(verbose=cfg.parser_verbose)

    # Parse all sentences (train + val) for sentence1 and sentence2
    all_s1 = pd.concat([train_df["sentence1"], val_df["sentence1"]], ignore_index=True).tolist()
    all_s2 = pd.concat([train_df["sentence2"], val_df["sentence2"]], ignore_index=True).tolist()

    diags1, fail1 = parse_sentence_list(all_s1, parser, tokeniser)
    diags2, fail2 = parse_sentence_list(all_s2, parser, tokeniser)

    # Build a combined dataframe back
    all_pairs_df = pd.concat([train_df, val_df], ignore_index=True).copy()
    all_pairs_df["diag1"] = diags1
    all_pairs_df["diag2"] = diags2

    # Drop rows with parsing failures
    before = len(all_pairs_df)
    all_pairs_df = all_pairs_df.dropna(subset=["diag1", "diag2"]).reset_index(drop=True)
    dropped = before - len(all_pairs_df)

    with (out_dir / "parsing_stats.json").open("w", encoding="utf-8") as f:
        json.dump(
            {
                "initial_rows": int(before),
                "kept_rows": int(len(all_pairs_df)),
                "dropped_rows_due_to_parsing": int(dropped),
                "fail_sentence1_count": int(len(fail1)),
                "fail_sentence2_count": int(len(fail2)),
                "fail_sentence1_examples": fail1[:10],
                "fail_sentence2_examples": fail2[:10],
            },
            f, ensure_ascii=False, indent=2
        )

    # Re-split (after dropping unparsable rows) with same val_size, stratified
    train_df2, val_df2 = train_test_split(
        all_pairs_df,
        test_size=cfg.val_size,
        random_state=cfg.seed,
        stratify=all_pairs_df["label"],
        shuffle=True,
    )
    train_df2 = train_df2.reset_index(drop=True)
    val_df2 = val_df2.reset_index(drop=True)

    # Unknown words rule from TRAIN diagrams only
    train_diagrams_for_vocab = (
        train_df2["diag1"].tolist() + train_df2["diag2"].tolist()
    )
    unk_rule = UnknownWordsRewriteRule.from_diagrams(
        diagrams=train_diagrams_for_vocab,
        min_freq=cfg.min_freq_unk,
        ignore_types=True
    )
    rewriter = Rewriter([unk_rule])

    # Save vocabulary for later inference
    vocab = sorted(list(unk_rule.vocabulary))
    with (out_dir / "unk_vocabulary.json").open("w", encoding="utf-8") as f:
        json.dump(
            {"min_freq": cfg.min_freq_unk, "ignore_types": True, "vocabulary": vocab},
            f, ensure_ascii=False, indent=2
        )

    # Simplify diagrams + build circuits
    remove_cups = RemoveCupsRewriter()
    ansatz = IQPAnsatz(
        {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1, AtomicType.PREPOSITIONAL_PHRASE:0},
        n_layers=cfg.n_layers,
        n_single_qubit_params=cfg.n_single_qubit_params
    )

    def diag_to_circuit(diag_obj: object) -> object:
        d = rewriter(diag_obj)
        d = remove_cups(d)
        return ansatz(d)

    for split_name, split_df in [("train", train_df2), ("val", val_df2)]:
        split_df["circ1"] = [diag_to_circuit(d) for d in tqdm(split_df["diag1"], desc=f"Circuits {split_name} s1", leave=False)]
        split_df["circ2"] = [diag_to_circuit(d) for d in tqdm(split_df["diag2"], desc=f"Circuits {split_name} s2", leave=False)]

    # Prepare datasets (lambeq Dataset)
    train_pairs = list(zip(train_df2["circ1"].tolist(), train_df2["circ2"].tolist()))
    val_pairs = list(zip(val_df2["circ1"].tolist(), val_df2["circ2"].tolist()))

    y_train = train_df2["label"].astype(float).tolist()
    y_val = val_df2["label"].astype(float).tolist()

    # Set lambeq numerical backend to PyTorch so Dataset yields torch tensors
    lambeq_set_backend("pytorch")

    train_dataset = Dataset(train_pairs, y_train, batch_size=cfg.batch_size, shuffle=True)
    val_dataset = Dataset(val_pairs, y_val, batch_size=cfg.batch_size, shuffle=False)

    # Build model from ALL circuits we will use (train+val) so circuit_map is complete
    all_circuits = list(train_df2["circ1"]) + list(train_df2["circ2"]) + list(val_df2["circ1"]) + list(val_df2["circ2"])

    backend_config = {"backend": "default.qubit"}  # local simulator

    model = PairRedundancyModel.from_diagrams(
        all_circuits,
        probabilities=True,
        normalize=True,
        backend_config=backend_config,
        hidden_dim=16,
        dropout=0.10,
    )
    model.initialise_weights()

    # Class imbalance handling
    n_pos = int(sum(y_train))
    n_neg = int(len(y_train) - n_pos)
    pos_weight_value = (n_neg / max(n_pos, 1)) if cfg.imbalance_strategy == "pos_weight" else 1.0
    pos_weight = torch.tensor([pos_weight_value], dtype=torch.float32)

    def loss_fn(logits: torch.Tensor, y: torch.Tensor) -> torch.Tensor:
        y = y.float().view(-1)
        return torch.nn.functional.binary_cross_entropy_with_logits(
            logits, y, pos_weight=pos_weight.to(logits.device)
        )

    def acc_fn(logits: torch.Tensor, y: torch.Tensor) -> float:
        y = y.float().view(-1)
        preds = (torch.sigmoid(logits) >= 0.5).float()
        return float((preds == y).float().mean().item())

    trainer = PytorchTrainer(
        model=model,
        loss_function=loss_fn,
        optimizer=torch.optim.Adam,
        learning_rate=cfg.learning_rate,
        epochs=cfg.epochs,
        evaluate_functions={"acc": acc_fn},
        evaluate_on_train=True,
        use_tensorboard=False,
        log_dir=str(out_dir),
        verbose="text",
        seed=cfg.seed
    )

    # Train with early stopping on validation accuracy
    trainer.fit(
        train_dataset,
        val_dataset,
        early_stopping_criterion="acc",
        early_stopping_interval=cfg.early_stopping_patience,
        minimize_criterion=False
    )

    # Save final checkpoint explicitly (even if best_model exists)
    final_ckpt = out_dir / "final_model.lt"
    model.save(final_ckpt.as_posix())

    # Prefer best_model.lt if it exists
    best_ckpt = out_dir / "best_model.lt"
    ckpt_to_use = best_ckpt if best_ckpt.exists() else final_ckpt

    # Load chosen checkpoint into the same model instance
    model.load(ckpt_to_use.as_posix())

    # Evaluate on validation set
    model.eval()
    with torch.no_grad():
        val_logits = model(val_pairs).detach().cpu().numpy()
        val_prob = 1.0 / (1.0 + np.exp(-val_logits))

    y_val_np = np.array(y_val, dtype=int)
    metrics = evaluate_binary(y_val_np, val_prob)

    with (out_dir / "metrics_val.json").open("w", encoding="utf-8") as f:
        json.dump(metrics, f, ensure_ascii=False, indent=2)

    print("\n=== Validation metrics (QNLP pair model) ===")
    print(metrics["classification_report"])
    print("Confusion matrix:\n", np.array(metrics["confusion_matrix"]))
    print("Saved checkpoint used for eval:", ckpt_to_use.as_posix())

    # Export predictions for train+val
    full_df = pd.concat([train_df2.assign(split="train"), val_df2.assign(split="val")], ignore_index=True)
    full_pairs = list(zip(full_df["circ1"].tolist(), full_df["circ2"].tolist()))
    y_true_full = full_df["label"].astype(int).to_numpy()

    with torch.no_grad():
        logits_full = model(full_pairs).detach().cpu().numpy()
        prob_full = 1.0 / (1.0 + np.exp(-logits_full))
        pred_full = (prob_full >= 0.5).astype(int)

    pred_out = pd.DataFrame({
        "sentence1": full_df["sentence1"].tolist(),
        "sentence2": full_df["sentence2"].tolist(),
        "label_true": y_true_full,
        "split": full_df["split"].tolist(),
        "prob_redundancy": prob_full,
        "label_pred": pred_full,
    })
    pred_out.to_csv(out_dir / "predictions_qnlp.csv", index=False)

    # Optional baseline: TF-IDF + Logistic Regression
    if cfg.run_logreg_baseline:
        train_texts = (train_df2["sentence1"] + " [SEP] " + train_df2["sentence2"]).tolist()
        val_texts = (val_df2["sentence1"] + " [SEP] " + val_df2["sentence2"]).tolist()
        y_train_lr = train_df2["label"].astype(int).to_numpy()
        y_val_lr = val_df2["label"].astype(int).to_numpy()

        vectorizer = TfidfVectorizer(ngram_range=(1, 2), max_features=5000)
        X_train = vectorizer.fit_transform(train_texts)
        X_val = vectorizer.transform(val_texts)

        clf = LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=cfg.seed
        )
        clf.fit(X_train, y_train_lr)

        val_pred = clf.predict(X_val)
        val_prob_lr = clf.predict_proba(X_val)[:, 1]

        baseline_metrics = evaluate_binary(y_val_lr, val_prob_lr)
        with (out_dir / "metrics_val_logreg.json").open("w", encoding="utf-8") as f:
            json.dump(baseline_metrics, f, ensure_ascii=False, indent=2)

        joblib_dump(vectorizer, out_dir / "tfidf_vectorizer.joblib")
        joblib_dump(clf, out_dir / "logreg_model.joblib")

        print("\n=== Validation metrics (LogReg baseline) ===")
        print(baseline_metrics["classification_report"])

    print("\nArtifacts written to:", out_dir.resolve())


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--csv_path", type=str, default="/content/exclusion.csv")
    parser.add_argument("--out_dir", type=str, default="/content/outputs/run_pair_exclusion")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--val_size", type=float, default=0.20)
    parser.add_argument("--min_freq_unk", type=int, default=2)
    parser.add_argument("--batch_size", type=int, default=8)
    parser.add_argument("--epochs", type=int, default=60)
    parser.add_argument("--learning_rate", type=float, default=0.02)
    parser.add_argument("--early_stopping_patience", type=int, default=10)
    parser.add_argument("--imbalance_strategy", type=str, default="pos_weight",
                        choices=["pos_weight", "oversample", "none"])
    parser.add_argument("--run_logreg_baseline", action="store_true")

    # 🔥 FIX pentru Jupyter / Colab
    args, _ = parser.parse_known_args()

    cfg = Config(**vars(args))
    main(cfg)
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
train_lambeq_pair_redundancy.py

End-to-end: CSV(sentence1,sentence2,label) -> lambeq QNLP pair classifier
- Reproducible shuffle
- 80/20 stratified split
- Tokenize (SpacyTokeniser) + parse (BobcatParser) -> diagrams
- UNK handling (UnknownWordsRewriteRule) + diagram simplification (RemoveCupsRewriter)
- IQPAnsatz -> quantum circuits
- Pair model (PennyLaneModel subclass): run 2 circuits, concat outputs, small NN head
- Train (PytorchTrainer), handle class imbalance (pos_weight OR oversampling)
- Evaluate (acc/prec/rec/F1) and export predictions + save model checkpoint

Tested design target: lambeq 0.5.x, Python 3.10+, CPU.
"""

import os
import json
import argparse
import random
from dataclasses import asdict, dataclass
from pathlib import Path
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import torch
from torch import nn

from tqdm import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    classification_report
)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from joblib import dump as joblib_dump

from lambeq import (
    BobcatParser,
    SpacyTokeniser,
    RemoveCupsRewriter,
    IQPAnsatz,
    AtomicType,
    UnknownWordsRewriteRule,
    Rewriter,
    Dataset,
    PennyLaneModel,
    PytorchTrainer,
)
from lambeq.backend.numerical_backend import set_backend as lambeq_set_backend


@dataclass
class Config:
    csv_path: str
    out_dir: str = "/content/outputs/run_pair_implication"
    seed: int = 42
    val_size: float = 0.20

    # Parsing / preprocessing
    min_freq_unk: int = 2
    parser_verbose: str = "suppress"  # 'suppress' or 'text'

    # Circuit / ansatz
    n_layers: int = 1
    n_single_qubit_params: int = 3

    # Training
    batch_size: int = 8
    epochs: int = 60
    learning_rate: float = 0.02
    early_stopping_patience: int = 10

    # Class imbalance handling
    imbalance_strategy: str = "pos_weight"  # 'pos_weight' or 'oversample' or 'none'

    # Baseline
    run_logreg_baseline: bool = True


def set_all_seeds(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def clean_text(x: object) -> Optional[str]:
    if not isinstance(x, str):
        return None
    s = " ".join(x.strip().split())
    return s if s else None


def load_and_shuffle_csv(csv_path: str, seed: int) -> pd.DataFrame:
    df = pd.read_csv(csv_path)

    required = {"sentence1", "sentence2", "label"}
    if not required.issubset(df.columns):
        raise ValueError(f"CSV must contain columns {sorted(required)}. Found: {list(df.columns)}")

    df = df.copy()
    df["sentence1"] = df["sentence1"].apply(clean_text)
    df["sentence2"] = df["sentence2"].apply(clean_text)

    df = df.dropna(subset=["sentence1", "sentence2", "label"]).reset_index(drop=True)

    # Validate labels
    df["label"] = df["label"].astype(int)
    if not set(df["label"].unique()).issubset({0, 1}):
        raise ValueError(f"Labels must be binary 0/1. Found: {sorted(df['label'].unique())}")

    # Reproducible shuffle
    df = df.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return df


def parse_sentence_list(
    sentences: List[str],
    parser: BobcatParser,
    tokeniser: SpacyTokeniser,
) -> Tuple[List[Optional[object]], List[Tuple[int, str]]]:
    diagrams: List[Optional[object]] = []
    failures: List[Tuple[int, str]] = []

    for i, s in enumerate(tqdm(sentences, desc="Parsing sentences", leave=False)):
        try:
            tokens = tokeniser.tokenise_sentence(s)
            diag = parser.sentence2diagram(tokens, tokenised=True)
            diagrams.append(diag)
        except Exception as e:
            diagrams.append(None)
            failures.append((i, repr(e)))

    return diagrams, failures


def oversample_train_df(train_df: pd.DataFrame, seed: int) -> pd.DataFrame:
    counts = train_df["label"].value_counts()
    if len(counts) < 2:
        return train_df

    maj_label = int(counts.idxmax())
    min_label = int(counts.idxmin())
    n_maj = int(counts[maj_label])
    n_min = int(counts[min_label])

    if n_min == n_maj:
        return train_df

    n_to_add = n_maj - n_min
    min_df = train_df[train_df["label"] == min_label]
    add_df = min_df.sample(n=n_to_add, replace=True, random_state=seed)

    out = pd.concat([train_df, add_df], ignore_index=True)
    out = out.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return out


class PairRedundancyModel(PennyLaneModel):
    """Hybrid model: two circuits -> concat probs -> small NN -> logit"""

    def __init__(self, hidden_dim: int = 16, dropout: float = 0.10, **kwargs):
        super().__init__(**kwargs)
        self.net = nn.Sequential(
            nn.Linear(4, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 1)  # logit
        )

    def forward(self, diagram_pairs: List[Tuple[object, object]]) -> torch.Tensor:
        # diagram_pairs is a list of tuples (circ1, circ2)
        first, second = zip(*diagram_pairs)
        out1 = self.get_diagram_output(list(first))   # shape [B,2] for 1-qubit prob
        out2 = self.get_diagram_output(list(second))  # shape [B,2]
        x = torch.cat([out1, out2], dim=1)            # shape [B,4]

        # map [0,1] -> [-1,1] (helps ReLU)
        x = 2.0 * (x - 0.5)

        logits = self.net(x).squeeze(-1)              # shape [B]
        return logits


def evaluate_binary(y_true: np.ndarray, prob_pos: np.ndarray) -> dict:
    y_pred = (prob_pos >= 0.5).astype(int)

    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, _support = precision_recall_fscore_support(
        y_true, y_pred, average="binary", pos_label=1, zero_division=0
    )
    cm = confusion_matrix(y_true, y_pred)

    return {
        "accuracy": float(acc),
        "precision_pos1": float(prec),
        "recall_pos1": float(rec),
        "f1_pos1": float(f1),
        "confusion_matrix": cm.tolist(),
        "classification_report": classification_report(
            y_true, y_pred, digits=4, zero_division=0
        ),
    }


def main(cfg: Config) -> None:
    os.environ["TOKENIZERS_PARALLELISM"] = "true"
    set_all_seeds(cfg.seed)

    out_dir = Path(cfg.out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    # Save config for reproducibility
    with (out_dir / "config.json").open("w", encoding="utf-8") as f:
        json.dump(asdict(cfg), f, ensure_ascii=False, indent=2)

    # Load and split
    df = load_and_shuffle_csv(cfg.csv_path, cfg.seed)
    train_df, val_df = train_test_split(
        df,
        test_size=cfg.val_size,
        random_state=cfg.seed,
        stratify=df["label"],
        shuffle=True,
    )
    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)

    # Optional: oversample to handle imbalance
    if cfg.imbalance_strategy == "oversample":
        train_df = oversample_train_df(train_df, cfg.seed)

    # Tokeniser + parser
    tokeniser = SpacyTokeniser()
    parser = BobcatParser(verbose=cfg.parser_verbose)

    # Parse all sentences (train + val) for sentence1 and sentence2
    all_s1 = pd.concat([train_df["sentence1"], val_df["sentence1"]], ignore_index=True).tolist()
    all_s2 = pd.concat([train_df["sentence2"], val_df["sentence2"]], ignore_index=True).tolist()

    diags1, fail1 = parse_sentence_list(all_s1, parser, tokeniser)
    diags2, fail2 = parse_sentence_list(all_s2, parser, tokeniser)

    # Build a combined dataframe back
    all_pairs_df = pd.concat([train_df, val_df], ignore_index=True).copy()
    all_pairs_df["diag1"] = diags1
    all_pairs_df["diag2"] = diags2

    # Drop rows with parsing failures
    before = len(all_pairs_df)
    all_pairs_df = all_pairs_df.dropna(subset=["diag1", "diag2"]).reset_index(drop=True)
    dropped = before - len(all_pairs_df)

    with (out_dir / "parsing_stats.json").open("w", encoding="utf-8") as f:
        json.dump(
            {
                "initial_rows": int(before),
                "kept_rows": int(len(all_pairs_df)),
                "dropped_rows_due_to_parsing": int(dropped),
                "fail_sentence1_count": int(len(fail1)),
                "fail_sentence2_count": int(len(fail2)),
                "fail_sentence1_examples": fail1[:10],
                "fail_sentence2_examples": fail2[:10],
            },
            f, ensure_ascii=False, indent=2
        )

    # Re-split (after dropping unparsable rows) with same val_size, stratified
    train_df2, val_df2 = train_test_split(
        all_pairs_df,
        test_size=cfg.val_size,
        random_state=cfg.seed,
        stratify=all_pairs_df["label"],
        shuffle=True,
    )
    train_df2 = train_df2.reset_index(drop=True)
    val_df2 = val_df2.reset_index(drop=True)

    # Unknown words rule from TRAIN diagrams only
    train_diagrams_for_vocab = (
        train_df2["diag1"].tolist() + train_df2["diag2"].tolist()
    )
    unk_rule = UnknownWordsRewriteRule.from_diagrams(
        diagrams=train_diagrams_for_vocab,
        min_freq=cfg.min_freq_unk,
        ignore_types=True
    )
    rewriter = Rewriter([unk_rule])

    # Save vocabulary for later inference
    vocab = sorted(list(unk_rule.vocabulary))
    with (out_dir / "unk_vocabulary.json").open("w", encoding="utf-8") as f:
        json.dump(
            {"min_freq": cfg.min_freq_unk, "ignore_types": True, "vocabulary": vocab},
            f, ensure_ascii=False, indent=2
        )

    # Simplify diagrams + build circuits
    remove_cups = RemoveCupsRewriter()
    ansatz = IQPAnsatz(
        {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1, AtomicType.PREPOSITIONAL_PHRASE:0},
        n_layers=cfg.n_layers,
        n_single_qubit_params=cfg.n_single_qubit_params
    )

    def diag_to_circuit(diag_obj: object) -> object:
        d = rewriter(diag_obj)
        d = remove_cups(d)
        return ansatz(d)

    for split_name, split_df in [("train", train_df2), ("val", val_df2)]:
        split_df["circ1"] = [diag_to_circuit(d) for d in tqdm(split_df["diag1"], desc=f"Circuits {split_name} s1", leave=False)]
        split_df["circ2"] = [diag_to_circuit(d) for d in tqdm(split_df["diag2"], desc=f"Circuits {split_name} s2", leave=False)]

    # Prepare datasets (lambeq Dataset)
    train_pairs = list(zip(train_df2["circ1"].tolist(), train_df2["circ2"].tolist()))
    val_pairs = list(zip(val_df2["circ1"].tolist(), val_df2["circ2"].tolist()))

    y_train = train_df2["label"].astype(float).tolist()
    y_val = val_df2["label"].astype(float).tolist()

    # Set lambeq numerical backend to PyTorch so Dataset yields torch tensors
    lambeq_set_backend("pytorch")

    train_dataset = Dataset(train_pairs, y_train, batch_size=cfg.batch_size, shuffle=True)
    val_dataset = Dataset(val_pairs, y_val, batch_size=cfg.batch_size, shuffle=False)

    # Build model from ALL circuits we will use (train+val) so circuit_map is complete
    all_circuits = list(train_df2["circ1"]) + list(train_df2["circ2"]) + list(val_df2["circ1"]) + list(val_df2["circ2"])

    backend_config = {"backend": "default.qubit"}  # local simulator

    model = PairRedundancyModel.from_diagrams(
        all_circuits,
        probabilities=True,
        normalize=True,
        backend_config=backend_config,
        hidden_dim=16,
        dropout=0.10,
    )
    model.initialise_weights()

    # Class imbalance handling
    n_pos = int(sum(y_train))
    n_neg = int(len(y_train) - n_pos)
    pos_weight_value = (n_neg / max(n_pos, 1)) if cfg.imbalance_strategy == "pos_weight" else 1.0
    pos_weight = torch.tensor([pos_weight_value], dtype=torch.float32)

    def loss_fn(logits: torch.Tensor, y: torch.Tensor) -> torch.Tensor:
        y = y.float().view(-1)
        return torch.nn.functional.binary_cross_entropy_with_logits(
            logits, y, pos_weight=pos_weight.to(logits.device)
        )

    def acc_fn(logits: torch.Tensor, y: torch.Tensor) -> float:
        y = y.float().view(-1)
        preds = (torch.sigmoid(logits) >= 0.5).float()
        return float((preds == y).float().mean().item())

    trainer = PytorchTrainer(
        model=model,
        loss_function=loss_fn,
        optimizer=torch.optim.Adam,
        learning_rate=cfg.learning_rate,
        epochs=cfg.epochs,
        evaluate_functions={"acc": acc_fn},
        evaluate_on_train=True,
        use_tensorboard=False,
        log_dir=str(out_dir),
        verbose="text",
        seed=cfg.seed
    )

    # Train with early stopping on validation accuracy
    trainer.fit(
        train_dataset,
        val_dataset,
        early_stopping_criterion="acc",
        early_stopping_interval=cfg.early_stopping_patience,
        minimize_criterion=False
    )

    # Save final checkpoint explicitly (even if best_model exists)
    final_ckpt = out_dir / "final_model.lt"
    model.save(final_ckpt.as_posix())

    # Prefer best_model.lt if it exists
    best_ckpt = out_dir / "best_model.lt"
    ckpt_to_use = best_ckpt if best_ckpt.exists() else final_ckpt

    # Load chosen checkpoint into the same model instance
    model.load(ckpt_to_use.as_posix())

    # Evaluate on validation set
    model.eval()
    with torch.no_grad():
        val_logits = model(val_pairs).detach().cpu().numpy()
        val_prob = 1.0 / (1.0 + np.exp(-val_logits))

    y_val_np = np.array(y_val, dtype=int)
    metrics = evaluate_binary(y_val_np, val_prob)

    with (out_dir / "metrics_val.json").open("w", encoding="utf-8") as f:
        json.dump(metrics, f, ensure_ascii=False, indent=2)

    print("\n=== Validation metrics (QNLP pair model) ===")
    print(metrics["classification_report"])
    print("Confusion matrix:\n", np.array(metrics["confusion_matrix"]))
    print("Saved checkpoint used for eval:", ckpt_to_use.as_posix())

    # Export predictions for train+val
    full_df = pd.concat([train_df2.assign(split="train"), val_df2.assign(split="val")], ignore_index=True)
    full_pairs = list(zip(full_df["circ1"].tolist(), full_df["circ2"].tolist()))
    y_true_full = full_df["label"].astype(int).to_numpy()

    with torch.no_grad():
        logits_full = model(full_pairs).detach().cpu().numpy()
        prob_full = 1.0 / (1.0 + np.exp(-logits_full))
        pred_full = (prob_full >= 0.5).astype(int)

    pred_out = pd.DataFrame({
        "sentence1": full_df["sentence1"].tolist(),
        "sentence2": full_df["sentence2"].tolist(),
        "label_true": y_true_full,
        "split": full_df["split"].tolist(),
        "prob_redundancy": prob_full,
        "label_pred": pred_full,
    })
    pred_out.to_csv(out_dir / "predictions_qnlp.csv", index=False)

    # Optional baseline: TF-IDF + Logistic Regression
    if cfg.run_logreg_baseline:
        train_texts = (train_df2["sentence1"] + " [SEP] " + train_df2["sentence2"]).tolist()
        val_texts = (val_df2["sentence1"] + " [SEP] " + val_df2["sentence2"]).tolist()
        y_train_lr = train_df2["label"].astype(int).to_numpy()
        y_val_lr = val_df2["label"].astype(int).to_numpy()

        vectorizer = TfidfVectorizer(ngram_range=(1, 2), max_features=5000)
        X_train = vectorizer.fit_transform(train_texts)
        X_val = vectorizer.transform(val_texts)

        clf = LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=cfg.seed
        )
        clf.fit(X_train, y_train_lr)

        val_pred = clf.predict(X_val)
        val_prob_lr = clf.predict_proba(X_val)[:, 1]

        baseline_metrics = evaluate_binary(y_val_lr, val_prob_lr)
        with (out_dir / "metrics_val_logreg.json").open("w", encoding="utf-8") as f:
            json.dump(baseline_metrics, f, ensure_ascii=False, indent=2)

        joblib_dump(vectorizer, out_dir / "tfidf_vectorizer.joblib")
        joblib_dump(clf, out_dir / "logreg_model.joblib")

        print("\n=== Validation metrics (LogReg baseline) ===")
        print(baseline_metrics["classification_report"])

    print("\nArtifacts written to:", out_dir.resolve())


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--csv_path", type=str, default="/content/DA_generarat.csv")
    parser.add_argument("--out_dir", type=str, default="/content/outputs/run_pair_implication")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--val_size", type=float, default=0.20)
    parser.add_argument("--min_freq_unk", type=int, default=2)
    parser.add_argument("--batch_size", type=int, default=8)
    parser.add_argument("--epochs", type=int, default=60)
    parser.add_argument("--learning_rate", type=float, default=0.02)
    parser.add_argument("--early_stopping_patience", type=int, default=10)
    parser.add_argument("--imbalance_strategy", type=str, default="pos_weight",
                        choices=["pos_weight", "oversample", "none"])
    parser.add_argument("--run_logreg_baseline", action="store_true")

    # 🔥 FIX pentru Jupyter / Colab
    args, _ = parser.parse_known_args()

    cfg = Config(**vars(args))
    main(cfg)
```

```python
#%%writefile train_lambeq_pair_quantumcase.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
Quantum case (pure): CSV(sentence1,sentence2,label) -> lambeq TketModel + QuantumTrainer
- local shot-based simulator: pytket.extensions.qiskit.AerBackend
- pair-to-single-circuit design: tensor + CX + Discard (default)
- reproducible shuffle + 80/20 stratified split
- tokenization (SpacyTokeniser) + parsing (BobcatParser)
- UNK handling (UnknownWordsRewriteRule) + RemoveCupsRewriter
- IQPAnsatz -> circuits
- BinaryCrossEntropyLoss / custom weighted CE (for imbalance)
- metrics: accuracy, precision, recall, F1 + confusion matrix
- saves best_model.lt (trainer) + final_model.lt (explicit) + predictions CSV
"""

import os
import json
import argparse
import random
from dataclasses import dataclass, asdict
from pathlib import Path
from typing import List, Optional, Tuple

import numpy as np
import pandas as pd

from tqdm import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix, classification_report

from lambeq import (
    BobcatParser,
    SpacyTokeniser,
    RemoveCupsRewriter,
    IQPAnsatz,
    AtomicType,
    UnknownWordsRewriteRule,
    Rewriter,
    Dataset,
    TketModel,
    QuantumTrainer,
    SPSAOptimizer,
    BinaryCrossEntropyLoss
)
from lambeq.backend.quantum import CX, Id, Discard, qubit


@dataclass
class Config:
    csv_path: str
    out_dir: str = "/content/outputs/run_quantum_case_implication"
    seed: int = 42
    val_size: float = 0.20

    # preprocessing
    min_freq_unk: int = 2
    parser_verbose: str = "suppress"

    # pair circuit design
    pair_design: str = "cx_discard"  # "cx_discard" | "concat_text"

    # ansatz
    n_layers: int = 1
    n_single_qubit_params: int = 3

    # quantum backend
    shots: int = 2048
    compilation_optim: int = 2  # backend.default_compilation_pass(2)

    # training
    batch_size: int = 32
    epochs: int = 80
    early_stopping_patience: int = 10

    # imbalance
    imbalance_strategy: str = "oversample"  # "oversample" | "weighted_loss" | "none"


def set_all_seeds(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)


def clean_text(x: object) -> Optional[str]:
    if not isinstance(x, str):
        return None
    s = " ".join(x.strip().split())
    return s if s else None


def load_and_shuffle_csv(csv_path: str, seed: int) -> pd.DataFrame:
    df = pd.read_csv(csv_path)
    required = {"sentence1", "sentence2", "label"}
    if not required.issubset(df.columns):
        raise ValueError(f"CSV must contain columns {sorted(required)}. Found: {list(df.columns)}")

    df = df.copy()
    df["sentence1"] = df["sentence1"].apply(clean_text)
    df["sentence2"] = df["sentence2"].apply(clean_text)
    df = df.dropna(subset=["sentence1", "sentence2", "label"]).reset_index(drop=True)

    df["label"] = df["label"].astype(int)
    if not set(df["label"].unique()).issubset({0, 1}):
        raise ValueError(f"Labels must be binary 0/1. Found: {sorted(df['label'].unique())}")

    # reproducible shuffle
    df = df.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return df


def oversample_train_df(train_df: pd.DataFrame, seed: int) -> pd.DataFrame:
    counts = train_df["label"].value_counts()
    if len(counts) < 2:
        return train_df
    maj = int(counts.idxmax())
    mino = int(counts.idxmin())
    n_maj = int(counts[maj])
    n_min = int(counts[mino])
    if n_min == n_maj:
        return train_df

    n_to_add = n_maj - n_min
    min_df = train_df[train_df["label"] == mino]
    add_df = min_df.sample(n=n_to_add, replace=True, random_state=seed)

    out = pd.concat([train_df, add_df], ignore_index=True)
    out = out.sample(frac=1.0, random_state=seed).reset_index(drop=True)
    return out


def parse_one(sentence: str, parser: BobcatParser, tokeniser: SpacyTokeniser):
    tokens = tokeniser.tokenise_sentence(sentence)
    return parser.sentence2diagram(tokens, tokenised=True)


def build_pair_circuit(c1, c2, design: str):
    # Expect each circuit to have exactly 1 output qubit (sentence wire).
    # If not, you must discard down to 1 qubit, otherwise BCE(2-class) won't match.
    if design == "cx_discard":
        if len(c1.cod) != 1 or len(c2.cod) != 1:
            raise ValueError(f"Expected 1 output qubit per sentence circuit, got len(c1.cod)={len(c1.cod)}, len(c2.cod)={len(c2.cod)}")
        # (c1 @ c2) has 2 output qubits -> entangle -> discard second -> 1 output qubit
        return (c1 @ c2) >> CX >> (Id(qubit) @ Discard())
    elif design == "concat_text":
        raise RuntimeError("concat_text design is handled earlier at text->diagram stage.")
    else:
        raise ValueError(f"Unknown pair design: {design}")


def to_one_hot(labels01: List[int]) -> List[List[float]]:
    # Class order: 0 -> [1,0], 1 -> [0,1]
    return [[1.0, 0.0] if y == 0 else [0.0, 1.0] for y in labels01]


def weighted_ce_loss_factory(w0: float, w1: float, eps: float = 1e-9):
    # y_pred: [B,2] prob dist; y_true: [B,2] one-hot
    weights = np.array([w0, w1], dtype=np.float64)

    def loss(y_pred: np.ndarray, y_true: np.ndarray) -> float:
        y_pred = np.clip(y_pred, eps, 1.0 - eps)
        # -sum_k w_k * y_true_k * log(p_k)
        per_ex = -np.sum(y_true * np.log(y_pred) * weights[None, :], axis=1)
        return float(np.mean(per_ex))

    return loss


def acc_argmax(y_hat: np.ndarray, y: np.ndarray) -> float:
    return float(np.mean(np.argmax(y_hat, axis=1) == np.argmax(y, axis=1)))


def main(cfg: Config) -> None:
    os.environ["TOKENIZERS_PARALLELISM"] = "true"
    set_all_seeds(cfg.seed)

    out_dir = Path(cfg.out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    with (out_dir / "config.json").open("w", encoding="utf-8") as f:
        json.dump(asdict(cfg), f, ensure_ascii=False, indent=2)

    df = load_and_shuffle_csv(cfg.csv_path, cfg.seed)

    train_df, val_df = train_test_split(
        df, test_size=cfg.val_size, random_state=cfg.seed, stratify=df["label"], shuffle=True
    )
    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)

    if cfg.imbalance_strategy == "oversample":
        train_df = oversample_train_df(train_df, cfg.seed)

    tokeniser = SpacyTokeniser()
    parser = BobcatParser(verbose=cfg.parser_verbose)

    remove_cups = RemoveCupsRewriter()

    # ======== design "concat_text": parse one diagram per example ========
    if cfg.pair_design == "concat_text":
        def join_pair(s1, s2):
            return f"{s1}. {s2}."
        train_text = [join_pair(a, b) for a, b in zip(train_df["sentence1"], train_df["sentence2"])]
        val_text   = [join_pair(a, b) for a, b in zip(val_df["sentence1"], val_df["sentence2"])]

        train_diags, val_diags = [], []
        for s in tqdm(train_text, desc="Parsing train (concat_text)"):
            try:
                train_diags.append(parse_one(s, parser, tokeniser))
            except Exception:
                train_diags.append(None)
        for s in tqdm(val_text, desc="Parsing val (concat_text)"):
            try:
                val_diags.append(parse_one(s, parser, tokeniser))
            except Exception:
                val_diags.append(None)

        train_df = train_df.assign(diag=train_diags).dropna(subset=["diag"]).reset_index(drop=True)
        val_df   = val_df.assign(diag=val_diags).dropna(subset=["diag"]).reset_index(drop=True)

        # UNK vocab from train diagrams
        unk_rule = UnknownWordsRewriteRule.from_diagrams(
            diagrams=train_df["diag"].tolist(),
            min_freq=cfg.min_freq_unk,
            ignore_types=True
        )
        rewriter = Rewriter([unk_rule])

        # ansatz: ensure SENTENCE has 1 qubit
        ansatz = IQPAnsatz(
            {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1, AtomicType.PREPOSITIONAL_PHRASE: 1},
            n_layers=cfg.n_layers,
            n_single_qubit_params=cfg.n_single_qubit_params
        )

        train_circuits = [ansatz(remove_cups(rewriter(d))) for d in tqdm(train_df["diag"], desc="Circuits train")]
        val_circuits   = [ansatz(remove_cups(rewriter(d))) for d in tqdm(val_df["diag"], desc="Circuits val")]

    # ======== design "cx_discard": parse 2 diags per example then fuse circuits ========
    else:
        # Parse sentences independently
        train_diag1, train_diag2 = [], []
        val_diag1, val_diag2 = [], []

        for s in tqdm(train_df["sentence1"].tolist(), desc="Parsing train sentence1"):
            try:
                train_diag1.append(parse_one(s, parser, tokeniser))
            except Exception:
                train_diag1.append(None)

        for s in tqdm(train_df["sentence2"].tolist(), desc="Parsing train sentence2"):
            try:
                train_diag2.append(parse_one(s, parser, tokeniser))
            except Exception:
                train_diag2.append(None)

        for s in tqdm(val_df["sentence1"].tolist(), desc="Parsing val sentence1"):
            try:
                val_diag1.append(parse_one(s, parser, tokeniser))
            except Exception:
                val_diag1.append(None)

        for s in tqdm(val_df["sentence2"].tolist(), desc="Parsing val sentence2"):
            try:
                val_diag2.append(parse_one(s, parser, tokeniser))
            except Exception:
                val_diag2.append(None)

        # Drop unparsable rows
        train_df = train_df.assign(diag1=train_diag1, diag2=train_diag2).dropna(subset=["diag1", "diag2"]).reset_index(drop=True)
        val_df   = val_df.assign(diag1=val_diag1,   diag2=val_diag2).dropna(subset=["diag1", "diag2"]).reset_index(drop=True)

        # UNK vocab from TRAIN diagrams only (from both columns)
        vocab_diags = train_df["diag1"].tolist() + train_df["diag2"].tolist()
        unk_rule = UnknownWordsRewriteRule.from_diagrams(
            diagrams=vocab_diags,
            min_freq=cfg.min_freq_unk,
            ignore_types=True
        )
        rewriter = Rewriter([unk_rule])

        # Save vocab for inference
        vocab = sorted(list(unk_rule.vocabulary))
        with (out_dir / "unk_vocabulary.json").open("w", encoding="utf-8") as f:
            json.dump({"min_freq": cfg.min_freq_unk, "ignore_types": True, "vocabulary": vocab}, f, ensure_ascii=False, indent=2)

        ansatz = IQPAnsatz(
            {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1, AtomicType.PREPOSITIONAL_PHRASE: 1},
            n_layers=cfg.n_layers,
            n_single_qubit_params=cfg.n_single_qubit_params
        )

        def diag_to_circ(d):
            return ansatz(remove_cups(rewriter(d)))

        # sentence circuits
        train_c1 = [diag_to_circ(d) for d in tqdm(train_df["diag1"], desc="Circuits train s1")]
        train_c2 = [diag_to_circ(d) for d in tqdm(train_df["diag2"], desc="Circuits train s2")]
        val_c1   = [diag_to_circ(d) for d in tqdm(val_df["diag1"], desc="Circuits val s1")]
        val_c2   = [diag_to_circ(d) for d in tqdm(val_df["diag2"], desc="Circuits val s2")]

        # fuse into single pair circuit
        train_circuits = [build_pair_circuit(a, b, cfg.pair_design) for a, b in zip(train_c1, train_c2)]
        val_circuits   = [build_pair_circuit(a, b, cfg.pair_design) for a, b in zip(val_c1,   val_c2)]

    # Labels one-hot for BCE (dense mode)
    y_train_01 = train_df["label"].astype(int).tolist()
    y_val_01   = val_df["label"].astype(int).tolist()
    train_labels = to_one_hot(y_train_01)
    val_labels   = to_one_hot(y_val_01)

    # Build backend + model
    from pytket.extensions.qiskit import AerBackend

    backend = AerBackend()
    backend_config = {
        "backend": backend,
        "compilation": backend.default_compilation_pass(cfg.compilation_optim),
        "shots": cfg.shots
    }

    all_circuits = train_circuits + val_circuits
    model = TketModel.from_diagrams(all_circuits, backend_config=backend_config)
    model.initialise_weights()
    # Make TketModel RNG more reproducible than default_rng()
    model.rng = np.random.default_rng(cfg.seed)

    # Loss: BCE standard sau weighted CE
    if cfg.imbalance_strategy == "weighted_loss":
        n_pos = int(sum(y_train_01))
        n_neg = int(len(y_train_01) - n_pos)
        w1 = (n_neg / max(n_pos, 1))
        w0 = 1.0
        loss_fn = weighted_ce_loss_factory(w0=w0, w1=w1)
        with (out_dir / "class_weights.json").open("w", encoding="utf-8") as f:
            json.dump({"w0": w0, "w1": float(w1), "n_pos": n_pos, "n_neg": n_neg}, f, ensure_ascii=False, indent=2)
    else:
        loss_fn = BinaryCrossEntropyLoss(sparse=False)

    # Training steps estimate for SPSA 'A'
    steps_per_epoch = int(np.ceil(len(train_circuits) / max(cfg.batch_size, 1)))
    total_steps = cfg.epochs * steps_per_epoch
    A = max(1.0, 0.01 * total_steps)  # per SPSA docs: ~0.01 * num training steps

    trainer = QuantumTrainer(
        model=model,
        loss_function=loss_fn,
        epochs=cfg.epochs,
        optimizer=SPSAOptimizer,
        optim_hyperparams={"a": 0.05, "c": 0.06, "A": float(A)},
        evaluate_functions={"acc": acc_argmax},
        evaluate_on_train=True,
        verbose="text",
        log_dir=str(out_dir),
        seed=cfg.seed
    )

    train_dataset = Dataset(train_circuits, train_labels, batch_size=cfg.batch_size, shuffle=True)
    val_dataset   = Dataset(val_circuits,   val_labels,   batch_size=cfg.batch_size, shuffle=False)

    trainer.fit(
        train_dataset,
        val_dataset,
        log_interval=1,
        eval_interval=1,
        early_stopping_criterion="acc",
        early_stopping_interval=cfg.early_stopping_patience,
        minimize_criterion=False
    )

    # Load best checkpoint (created by trainer)
    best_ckpt = Path(trainer.log_dir) / "best_model.lt"
    if best_ckpt.exists():
        model.load(best_ckpt.as_posix())

    # Save a final explicit copy
    model.save((out_dir / "final_model.lt").as_posix())

    # Evaluate validation metrics
    y_hat = model(val_circuits)  # [N,2] probs
    y_true = np.argmax(np.array(val_labels), axis=1)
    y_pred = np.argmax(y_hat, axis=1)
    prob_pos = y_hat[:, 1]

    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, _ = precision_recall_fscore_support(y_true, y_pred, average="binary", pos_label=1, zero_division=0)
    cm = confusion_matrix(y_true, y_pred)

    metrics = {
        "accuracy": float(acc),
        "precision_pos1": float(prec),
        "recall_pos1": float(rec),
        "f1_pos1": float(f1),
        "confusion_matrix": cm.tolist(),
        "classification_report": classification_report(y_true, y_pred, digits=4, zero_division=0),
        "shots": cfg.shots
    }
    with (out_dir / "metrics_val.json").open("w", encoding="utf-8") as f:
        json.dump(metrics, f, ensure_ascii=False, indent=2)

    print("\n=== Validation metrics (Quantum case: TketModel + QuantumTrainer) ===")
    print(metrics["classification_report"])
    print("Confusion matrix:\n", np.array(metrics["confusion_matrix"]))
    print("best_model.lt:", best_ckpt.as_posix() if best_ckpt.exists() else "(not found)")
    print("final_model.lt:", (out_dir / "final_model.lt").as_posix())

    # Export predictions
    pred_df = pd.DataFrame({
        "sentence1": val_df["sentence1"].tolist(),
        "sentence2": val_df["sentence2"].tolist(),
        "label_true": y_true,
        "label_pred": y_pred,
        "prob_pos1": prob_pos
    })
    pred_df.to_csv(out_dir / "predictions_val_quantumcase.csv", index=False)
    print("\nArtifacts:", out_dir.resolve())


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--csv_path", type=str, default="/content/DA_generarat.csv")
    parser.add_argument("--out_dir", type=str, default="/content/outputs/run_quantum_case_implication")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--val_size", type=float, default=0.20)
    parser.add_argument("--min_freq_unk", type=int, default=2)
    parser.add_argument("--pair_design", type=str, default="cx_discard", choices=["cx_discard", "concat_text"])
    parser.add_argument("--n_layers", type=int, default=1)
    parser.add_argument("--n_single_qubit_params", type=int, default=3)
    parser.add_argument("--shots", type=int, default=2048)
    parser.add_argument("--batch_size", type=int, default=32)
    parser.add_argument("--epochs", type=int, default=80)
    parser.add_argument("--early_stopping_patience", type=int, default=10)
    parser.add_argument("--imbalance_strategy", type=str, default="oversample", choices=["oversample", "weighted_loss", "none"])

    # FIX Colab/Jupyter: ignoră argumentele interne (-f kernel.json)
    args, _ = parser.parse_known_args()
    cfg = Config(**vars(args))
    main(cfg)
```

```python
#%%writefile train_lambeq_pair_quantumcase_safe.py
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
Quantum case (pure): CSV(sentence1,sentence2,label) -> lambeq TketModel + QuantumTrainer
SAFE version for Colab:
- handles post_selection -> avoids tk converter tuple index errors
- handles None diagrams/circuits -> drops problematic rows
- runs on local shot-based simulator (pytket-qiskit AerBackend)

Input CSV columns: sentence1, sentence2, label (0/1)
"""

import os
import json
import argparse
import random
from dataclasses import dataclass, asdict
from pathlib import Path
from typing import List, Optional

import numpy as np
import pandas as pd
from tqdm import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix, classification_report

from lambeq import (
    BobcatParser,
    SpacyTokeniser,
    RemoveCupsRewriter,
    IQPAnsatz,
    AtomicType,
    UnknownWordsRewriteRule,
    Rewriter,
    Dataset,
    TketModel,
    QuantumTrainer,
    SPSAOptimizer,
    BinaryCrossEntropyLoss,
    CurryRewriteRule,
)
from lambeq.backend.quantum import CX, Id, Discard, qubit


# ---------------- CONFIG ----------------

@dataclass
class Config:
    csv_path: str
    out_dir: str = "/content/outputs/run_quantum_case_safe"
    seed: int = 42
    val_size: float = 0.20

    # preprocessing
    min_freq_unk: int = 2
    parser_verbose: str = "suppress"

    # pair circuit design
    pair_design: str = "cx_discard"  # "cx_discard" | "concat_text"

    # ansatz
    n_layers: int = 1
    n_single_qubit_params: int = 3

    # quantum backend
    shots: int = 1024
    compilation_optim: int = 2

    # training
    batch_size: int = 16
    epochs: int = 60
    early_stopping_patience: int = 10

    # imbalance
    imbalance_strategy: str = "oversample"  # "oversample" | "weighted_loss" | "none"


# ---------------- UTILS ----------------

def set_all_seeds(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)


def clean_text(x: object) -> Optional[str]:
    if not isinstance(x, str):
        return None
    s = " ".join(x.strip().split())
    return s if s else None


def load_and_shuffle_csv(csv_path: str, seed: int) -> pd.DataFrame:
    df = pd.read_csv(csv_path)
    required = {"sentence1", "sentence2", "label"}
    if not required.issubset(df.columns):
        raise ValueError(f"CSV must contain columns {sorted(required)}. Found: {list(df.columns)}")

    df = df.copy()
    df["sentence1"] = df["sentence1"].apply(clean_text)
    df["sentence2"] = df["sentence2"].apply(clean_text)
    df = df.dropna(subset=["sentence1", "sentence2", "label"]).reset_index(drop=True)

    df["label"] = df["label"].astype(int)
    if not set(df["label"].unique()).issubset({0, 1}):
        raise ValueError(f"Labels must be binary 0/1. Found: {sorted(df['label'].unique())}")

    return df.sample(frac=1.0, random_state=seed).reset_index(drop=True)


def oversample_train_df(train_df: pd.DataFrame, seed: int) -> pd.DataFrame:
    counts = train_df["label"].value_counts()
    if len(counts) < 2:
        return train_df

    maj = int(counts.idxmax())
    mino = int(counts.idxmin())
    n_maj = int(counts[maj])
    n_min = int(counts[mino])

    if n_min == n_maj:
        return train_df

    n_to_add = n_maj - n_min
    min_df = train_df[train_df["label"] == mino]
    add_df = min_df.sample(n=n_to_add, replace=True, random_state=seed)

    out = pd.concat([train_df, add_df], ignore_index=True)
    return out.sample(frac=1.0, random_state=seed).reset_index(drop=True)


def parse_one(sentence: str, parser: BobcatParser, tokeniser: SpacyTokeniser):
    tokens = tokeniser.tokenise_sentence(sentence)
    return parser.sentence2diagram(tokens, tokenised=True)


def build_pair_circuit(c1, c2, design: str):
    # Expect each circuit has 1 output qubit (sentence wire)
    if design == "cx_discard":
        if len(c1.cod) != 1 or len(c2.cod) != 1:
            raise ValueError(
                f"Expected 1 output qubit per sentence circuit, got len(c1.cod)={len(c1.cod)}, len(c2.cod)={len(c2.cod)}"
            )
        return (c1 @ c2) >> CX >> (Id(qubit) @ Discard())
    elif design == "concat_text":
        raise RuntimeError("concat_text handled earlier at text stage.")
    else:
        raise ValueError(f"Unknown pair design: {design}")


def to_one_hot(labels01: List[int]) -> List[List[float]]:
    # class 0 -> [1,0], class 1 -> [0,1]
    return [[1.0, 0.0] if y == 0 else [0.0, 1.0] for y in labels01]


def weighted_ce_loss_factory(w0: float, w1: float, eps: float = 1e-9):
    weights = np.array([w0, w1], dtype=np.float64)

    def loss(y_pred: np.ndarray, y_true: np.ndarray) -> float:
        y_pred2 = np.clip(y_pred, eps, 1.0 - eps)
        per_ex = -np.sum(y_true * np.log(y_pred2) * weights[None, :], axis=1)
        return float(np.mean(per_ex))

    return loss


def acc_argmax(y_hat: np.ndarray, y: np.ndarray) -> float:
    return float(np.mean(np.argmax(y_hat, axis=1) == np.argmax(y, axis=1)))


# ---------------- MAIN ----------------

def main(cfg: Config) -> None:
    os.environ["TOKENIZERS_PARALLELISM"] = "true"
    set_all_seeds(cfg.seed)

    out_dir = Path(cfg.out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    with (out_dir / "config.json").open("w", encoding="utf-8") as f:
        json.dump(asdict(cfg), f, ensure_ascii=False, indent=2)

    df = load_and_shuffle_csv(cfg.csv_path, cfg.seed)

    train_df, val_df = train_test_split(
        df, test_size=cfg.val_size, random_state=cfg.seed, stratify=df["label"], shuffle=True
    )
    train_df = train_df.reset_index(drop=True)
    val_df = val_df.reset_index(drop=True)

    if cfg.imbalance_strategy == "oversample":
        train_df = oversample_train_df(train_df, cfg.seed)

    tokeniser = SpacyTokeniser()
    parser = BobcatParser(verbose=cfg.parser_verbose)

    remove_cups = RemoveCupsRewriter()
    curry = CurryRewriteRule()

    # ✅ Keep SENTENCE=1 qubit; PP=0 reduces weirdness
    ansatz = IQPAnsatz(
        {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1, AtomicType.PREPOSITIONAL_PHRASE: 0},
        n_layers=cfg.n_layers,
        n_single_qubit_params=cfg.n_single_qubit_params
    )

    # ---------------- concat_text option ----------------
    if cfg.pair_design == "concat_text":
        def join_pair(a, b):
            return f"{a}. {b}."

        train_text = [join_pair(a, b) for a, b in zip(train_df["sentence1"], train_df["sentence2"])]
        val_text = [join_pair(a, b) for a, b in zip(val_df["sentence1"], val_df["sentence2"])]

        train_diags, val_diags = [], []
        for s in tqdm(train_text, desc="Parsing train (concat_text)"):
            try:
                train_diags.append(parse_one(s, parser, tokeniser))
            except Exception:
                train_diags.append(None)
        for s in tqdm(val_text, desc="Parsing val (concat_text)"):
            try:
                val_diags.append(parse_one(s, parser, tokeniser))
            except Exception:
                val_diags.append(None)

        train_df = train_df.assign(diag=train_diags).dropna(subset=["diag"]).reset_index(drop=True)
        val_df = val_df.assign(diag=val_diags).dropna(subset=["diag"]).reset_index(drop=True)

        # UNK vocab from train only
        unk_rule = UnknownWordsRewriteRule.from_diagrams(
            diagrams=train_df["diag"].tolist(),
            min_freq=cfg.min_freq_unk,
            ignore_types=True
        )
        rewriter = Rewriter([unk_rule])

        def diag_to_circ_safe(d):
            try:
                if d is None:
                    return None
                d2 = rewriter(d)
                if d2 is None:
                    return None
                d2 = curry(d2)
                if d2 is None:
                    return None
                d2 = d2.normal_form()
                d2 = remove_cups(d2)
                circ = ansatz(d2)
                if hasattr(circ, "post_selection") and isinstance(circ.post_selection, dict) and len(circ.post_selection) > 0:
                    circ.post_selection = {}
                return circ
            except Exception:
                return None

        train_circuits = [diag_to_circ_safe(d) for d in tqdm(train_df["diag"], desc="Circuits train")]
        val_circuits = [diag_to_circ_safe(d) for d in tqdm(val_df["diag"], desc="Circuits val")]

        train_df = train_df.assign(c=train_circuits).dropna(subset=["c"]).reset_index(drop=True)
        val_df = val_df.assign(c=val_circuits).dropna(subset=["c"]).reset_index(drop=True)

        train_circuits = train_df["c"].tolist()
        val_circuits = val_df["c"].tolist()

    # ---------------- cx_discard option (recommended) ----------------
    else:
        # Parse separately
        train_diag1, train_diag2 = [], []
        val_diag1, val_diag2 = [], []

        for s in tqdm(train_df["sentence1"].tolist(), desc="Parsing train sentence1"):
            try:
                train_diag1.append(parse_one(s, parser, tokeniser))
            except Exception:
                train_diag1.append(None)

        for s in tqdm(train_df["sentence2"].tolist(), desc="Parsing train sentence2"):
            try:
                train_diag2.append(parse_one(s, parser, tokeniser))
            except Exception:
                train_diag2.append(None)

        for s in tqdm(val_df["sentence1"].tolist(), desc="Parsing val sentence1"):
            try:
                val_diag1.append(parse_one(s, parser, tokeniser))
            except Exception:
                val_diag1.append(None)

        for s in tqdm(val_df["sentence2"].tolist(), desc="Parsing val sentence2"):
            try:
                val_diag2.append(parse_one(s, parser, tokeniser))
            except Exception:
                val_diag2.append(None)

        train_df = train_df.assign(diag1=train_diag1, diag2=train_diag2).dropna(subset=["diag1", "diag2"]).reset_index(drop=True)
        val_df = val_df.assign(diag1=val_diag1, diag2=val_diag2).dropna(subset=["diag1", "diag2"]).reset_index(drop=True)

        # UNK from train only (both columns)
        vocab_diags = train_df["diag1"].tolist() + train_df["diag2"].tolist()
        unk_rule = UnknownWordsRewriteRule.from_diagrams(
            diagrams=vocab_diags,
            min_freq=cfg.min_freq_unk,
            ignore_types=True
        )
        rewriter = Rewriter([unk_rule])

        with (out_dir / "unk_vocabulary.json").open("w", encoding="utf-8") as f:
            json.dump(
                {"min_freq": cfg.min_freq_unk, "ignore_types": True, "vocabulary": sorted(list(unk_rule.vocabulary))},
                f, ensure_ascii=False, indent=2
            )

        def diag_to_circ_safe(d):
            try:
                if d is None:
                    return None
                d2 = rewriter(d)
                if d2 is None:
                    return None
                d2 = curry(d2)
                if d2 is None:
                    return None
                d2 = d2.normal_form()
                d2 = remove_cups(d2)
                circ = ansatz(d2)
                if hasattr(circ, "post_selection") and isinstance(circ.post_selection, dict) and len(circ.post_selection) > 0:
                    circ.post_selection = {}
                return circ
            except Exception:
                return None

        # Build sentence circuits
        train_c1 = [diag_to_circ_safe(d) for d in tqdm(train_df["diag1"], desc="Circuits train s1")]
        train_c2 = [diag_to_circ_safe(d) for d in tqdm(train_df["diag2"], desc="Circuits train s2")]
        train_df = train_df.assign(c1=train_c1, c2=train_c2).dropna(subset=["c1", "c2"]).reset_index(drop=True)

        val_c1 = [diag_to_circ_safe(d) for d in tqdm(val_df["diag1"], desc="Circuits val s1")]
        val_c2 = [diag_to_circ_safe(d) for d in tqdm(val_df["diag2"], desc="Circuits val s2")]
        val_df = val_df.assign(c1=val_c1, c2=val_c2).dropna(subset=["c1", "c2"]).reset_index(drop=True)

        print("Kept after circuit build -> train:", len(train_df), "val:", len(val_df))

        # Fuse pair circuits
        train_circuits = [build_pair_circuit(a, b, cfg.pair_design) for a, b in zip(train_df["c1"], train_df["c2"])]
        val_circuits = [build_pair_circuit(a, b, cfg.pair_design) for a, b in zip(val_df["c1"], val_df["c2"])]

    # Labels AFTER dropping rows
    y_train_01 = train_df["label"].astype(int).tolist()
    y_val_01 = val_df["label"].astype(int).tolist()
    train_labels = to_one_hot(y_train_01)
    val_labels = to_one_hot(y_val_01)

    # Backend + model
    from pytket.extensions.qiskit import AerBackend
    backend = AerBackend()
    backend_config = {
        "backend": backend,
        "compilation": backend.default_compilation_pass(cfg.compilation_optim),
        "shots": cfg.shots
    }

    all_circuits = train_circuits + val_circuits
    model = TketModel.from_diagrams(all_circuits, backend_config=backend_config)
    model.initialise_weights()
    model.rng = np.random.default_rng(cfg.seed)

    # Loss
    if cfg.imbalance_strategy == "weighted_loss":
        n_pos = int(sum(y_train_01))
        n_neg = int(len(y_train_01) - n_pos)
        w1 = (n_neg / max(n_pos, 1))
        w0 = 1.0
        loss_fn = weighted_ce_loss_factory(w0=w0, w1=w1)
        with (out_dir / "class_weights.json").open("w", encoding="utf-8") as f:
            json.dump({"w0": w0, "w1": float(w1), "n_pos": n_pos, "n_neg": n_neg}, f, ensure_ascii=False, indent=2)
    else:
        loss_fn = BinaryCrossEntropyLoss(sparse=False)

    # SPSA A parameter
    steps_per_epoch = int(np.ceil(len(train_circuits) / max(cfg.batch_size, 1)))
    total_steps = cfg.epochs * steps_per_epoch
    A = max(1.0, 0.01 * total_steps)

    trainer = QuantumTrainer(
        model=model,
        loss_function=loss_fn,
        epochs=cfg.epochs,
        optimizer=SPSAOptimizer,
        optim_hyperparams={"a": 0.05, "c": 0.06, "A": float(A)},
        evaluate_functions={"acc": acc_argmax},
        evaluate_on_train=True,
        verbose="text",
        log_dir=str(out_dir),
        seed=cfg.seed
    )

    train_dataset = Dataset(train_circuits, train_labels, batch_size=cfg.batch_size, shuffle=True)
    val_dataset = Dataset(val_circuits, val_labels, batch_size=cfg.batch_size, shuffle=False)

    trainer.fit(
        train_dataset,
        val_dataset,
        log_interval=1,
        eval_interval=1,
        early_stopping_criterion="acc",
        early_stopping_interval=cfg.early_stopping_patience,
        minimize_criterion=False
    )

    # Load best
    best_ckpt = Path(trainer.log_dir) / "best_model.lt"
    if best_ckpt.exists():
        model.load(best_ckpt.as_posix())

    # Save final
    model.save((out_dir / "final_model.lt").as_posix())

    # Evaluate on val
    y_hat = model(val_circuits)  # [N,2]
    y_true = np.argmax(np.array(val_labels), axis=1)
    y_pred = np.argmax(y_hat, axis=1)
    prob_pos = y_hat[:, 1]

    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, _ = precision_recall_fscore_support(y_true, y_pred, average="binary", pos_label=1, zero_division=0)
    cm = confusion_matrix(y_true, y_pred)

    metrics = {
        "accuracy": float(acc),
        "precision_pos1": float(prec),
        "recall_pos1": float(rec),
        "f1_pos1": float(f1),
        "confusion_matrix": cm.tolist(),
        "classification_report": classification_report(y_true, y_pred, digits=4, zero_division=0),
        "shots": cfg.shots,
        "kept_train": int(len(train_df)),
        "kept_val": int(len(val_df)),
    }
    with (out_dir / "metrics_val.json").open("w", encoding="utf-8") as f:
        json.dump(metrics, f, ensure_ascii=False, indent=2)

    print("\n=== Validation metrics (Quantum SAFE: TketModel + QuantumTrainer) ===")
    print(metrics["classification_report"])
    print("Confusion matrix:\n", np.array(metrics["confusion_matrix"]))
    print("best_model.lt:", best_ckpt.as_posix() if best_ckpt.exists() else "(not found)")
    print("final_model.lt:", (out_dir / "final_model.lt").as_posix())
    print("Kept train/val:", metrics["kept_train"], metrics["kept_val"])
    print("Artifacts:", out_dir.resolve())

    # Export val predictions
    pred_df = pd.DataFrame({
        "sentence1": val_df["sentence1"].tolist(),
        "sentence2": val_df["sentence2"].tolist(),
        "label_true": y_true,
        "label_pred": y_pred,
        "prob_pos1": prob_pos,
    })
    pred_df.to_csv(out_dir / "predictions_val_quantumcase.csv", index=False)


if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--csv_path", type=str, default="/content/DA_generarat.csv")
    parser.add_argument("--out_dir", type=str, default="/content/outputs/run_quantum_case_safe")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--val_size", type=float, default=0.20)
    parser.add_argument("--min_freq_unk", type=int, default=2)
    parser.add_argument("--pair_design", type=str, default="cx_discard", choices=["cx_discard", "concat_text"])
    parser.add_argument("--n_layers", type=int, default=1)
    parser.add_argument("--n_single_qubit_params", type=int, default=3)
    parser.add_argument("--shots", type=int, default=1024)
    parser.add_argument("--batch_size", type=int, default=16)
    parser.add_argument("--epochs", type=int, default=60)
    parser.add_argument("--early_stopping_patience", type=int, default=10)
    parser.add_argument("--imbalance_strategy", type=str, default="oversample",
                        choices=["oversample", "weighted_loss", "none"])

    # Colab/Jupyter safe
    args, _ = parser.parse_known_args()
    cfg = Config(**vars(args))
    main(cfg)
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
Binary relation classification (neutral vs implication) on sentence pairs using lambeq + QuantumTrainer.

CSV format:
    sentence1,sentence2,label
where label ∈ {0,1}:
    0 => neutral
    1 => implication

Default backend: AerBackend (local).
This script also explains how to switch backends, but cloud Quantinuum runs are now primarily via qnexus/Nexus.

Tested logic targets lambeq-style pipeline:
tokenize -> parse (BobcatParser) -> DisCoCat diagrams -> RemoveCupsRewriter -> IQPAnsatz -> TketModel -> QuantumTrainer(SPSA)
"""

import argparse
import os
import warnings
from typing import List, Optional, Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

from lambeq import (
    BobcatParser,
    SpacyTokeniser,
    RemoveCupsRewriter,
    AtomicType,
    IQPAnsatz,
    TketModel,
    QuantumTrainer,
    SPSAOptimizer,
    BinaryCrossEntropyLoss,
    Dataset,
)

from lambeq.backend.symbol import Symbol
from lambeq.backend.quantum import (
    qubit,
    Id,
    CX,
    Rx,
    Rz,
    Discard,
)


CLASS_NAMES = ["neutral", "implication"]  # index 0/1 in our encoding


def read_csv_pairs(path: str) -> pd.DataFrame:
    if not os.path.exists(path):
        raise FileNotFoundError(f"CSV not found: {path}")

    df = pd.read_csv(path)
    required = {"sentence1", "sentence2", "label"}
    missing = required - set(df.columns)
    if missing:
        raise ValueError(
            f"CSV missing required columns: {sorted(missing)}. Expected columns: {sorted(required)}"
        )

    df = df.dropna(subset=["sentence1", "sentence2", "label"]).copy()
    df["label"] = df["label"].astype(int)

    bad = df[~df["label"].isin([0, 1])]
    if len(bad) > 0:
        raise ValueError(
            f"Found labels outside {{0,1}}. Examples:\n{bad[['label']].head()}"
        )

    # Basic cleanup (keep conservative; avoid heavy normalization unless needed)
    df["sentence1"] = df["sentence1"].astype(str).str.strip()
    df["sentence2"] = df["sentence2"].astype(str).str.strip()
    return df


def label_to_onehot(y01: int) -> List[int]:
    # Follow the common lambeq tutorial convention: 2D one-hot vectors
    # (class 0 => [1,0], class 1 => [0,1])
    return [1, 0] if y01 == 0 else [0, 1]


def parse_sentences(
    sentences: List[str],
    tokeniser: SpacyTokeniser,
    parser: BobcatParser,
    verbose: bool = False,
) -> List[Optional[object]]:
    """
    Tokenize with SpacyTokeniser, then parse each token list with BobcatParser.
    Return list of diagrams (or None if parsing fails).
    """
    tok = tokeniser.tokenise_sentences(sentences)
    diagrams: List[Optional[object]] = []

    for i, tokens in enumerate(tok):
        try:
            d = parser.sentence2diagram(tokens, tokenised=True)
            if d is None:
                diagrams.append(None)
            else:
                diagrams.append(d.normal_form())  # simplification, per tutorial patterns
        except Exception as e:
            if verbose:
                print(f"[parse] failed at index={i} tokens={tokens} err={type(e).__name__}: {e}")
            diagrams.append(None)

    return diagrams


def build_pair_head(n_params: int = 4):
    """
    A small trainable 'pair head' that:
        - acts on two qubits (one per sentence circuit),
        - entangles them (CX),
        - applies parameterized local rotations,
        - discards the second qubit,
        - leaves exactly one output qubit -> output probs [p0, p1].
    """
    if n_params < 2:
        raise ValueError("n_params must be >= 2")

    # Global/shared parameters (same for all datapoints)
    syms = [Symbol(f"rel_{i}") for i in range(n_params)]

    # Note: lambeq rotation parameters are in [0,1] (not [0, 2π]).
    # Keep the head shallow to avoid blowing up circuit depth.
    head = (Rz(syms[0]) @ Rz(syms[1])) >> CX

    # add more expressivity if desired
    if n_params >= 4:
        head = head >> (Rx(syms[2]) @ Rx(syms[3]))

    # compress 2 qubits -> 1 qubit
    head = head >> (Id(qubit) @ Discard())
    return head


def diagrams_to_pair_circuits(
    d1_list,
    d2_list,
    ansatz: IQPAnsatz,
    rewriter: RemoveCupsRewriter,
    pair_head,
    labels01: List[int],
    verbose: bool = False,
) -> Tuple[List[object], List[List[int]]]:
    circuits: List[object] = []
    labels_oh: List[List[int]] = []

    dropped = 0
    for i, (d1, d2, y) in enumerate(zip(d1_list, d2_list, labels01)):
        if d1 is None or d2 is None:
            dropped += 1
            continue
        try:
            c1 = ansatz(rewriter(d1))
            c2 = ansatz(rewriter(d2))
            pair_circ = (c1 @ c2) >> pair_head
            circuits.append(pair_circ)
            labels_oh.append(label_to_onehot(y))
        except Exception as e:
            dropped += 1
            if verbose:
                print(f"[circuit] failed at index={i} err={type(e).__name__}: {e}")

    if verbose:
        print(f"[circuit] kept={len(circuits)} dropped={dropped} (parse/circuit errors)")

    return circuits, labels_oh


def make_backend_config(backend: str, shots: int):
    """
    Backend options:
      - aer: local Qiskit Aer simulator via pytket-qiskit (recommended default)
      - quantinuum_local: local Quantinuum emulator (H2-1LE) via pytket-quantinuum[pecos] (if installed)

    NOTE: For cloud Quantinuum execution, current guidance is Nexus/`qnexus`.
    """
    backend = backend.lower().strip()

    if backend == "aer":
        from pytket.extensions.qiskit import AerBackend
        b = AerBackend()
        return {
            "backend": b,
            "compilation": b.default_compilation_pass(2),
            "shots": shots,
        }

    if backend == "quantinuum_local":
        # Requires: pip install pytket-quantinuum[pecos]
        # Local emulator device_name typically has suffix "LE", like "H2-1LE".
        from pytket.extensions.quantinuum import QuantinuumBackend
        b = QuantinuumBackend(device_name="H2-1LE")
        # Depending on environment/config, login may still be required (or offline API options).
        try:
            b.login()
        except Exception as e:
            raise RuntimeError(
                "QuantinuumBackend.login() failed. "
                "If you cannot authenticate, use --backend aer for local simulation."
            ) from e

        return {
            "backend": b,
            "compilation": b.default_compilation_pass(2),
            "shots": shots,
        }

    raise ValueError(f"Unknown backend '{backend}'. Use 'aer' or 'quantinuum_local'.")


def plot_training_curves(trainer, out_path: Optional[str] = None):
    """
    Plot loss + accuracy curves using QuantumTrainer stored fields
    (same interface as in official tutorial plots).
    """
    epochs = len(trainer.train_epoch_costs)
    xs = np.arange(1, epochs + 1)

    fig, ax1 = plt.subplots(figsize=(10, 5))
    ax1.set_xlabel("Epoch")
    ax1.set_ylabel("Loss")
    ax1.plot(xs, trainer.train_epoch_costs, label="train_loss")
    ax1.plot(xs, trainer.val_costs, label="val_loss")

    ax2 = ax1.twinx()
    ax2.set_ylabel("Accuracy")
    ax2.plot(xs, trainer.train_eval_results["acc"], label="train_acc")
    ax2.plot(xs, trainer.val_eval_results["acc"], label="val_acc")

    # Merge legends
    lines, labels = [], []
    for ax in (ax1, ax2):
        lns, lbs = ax.get_legend_handles_labels()
        lines.extend(lns)
        labels.extend(lbs)
    ax1.legend(lines, labels, loc="best")

    fig.tight_layout()
    if out_path:
        fig.savefig(out_path, dpi=150)
    else:
        plt.show()


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("--csv", type=str, default="/content/DA_generarat.csv", help="Path to CSV with sentence1,sentence2,label")
    parser.add_argument("--seed", type=int, default=2)
    parser.add_argument("--batch_size", type=int, default=16)
    parser.add_argument("--epochs", type=int, default=50)
    parser.add_argument("--shots", type=int, default=2048)
    parser.add_argument("--backend", type=str, default="aer", help="aer | quantinuum_local")
    parser.add_argument("--verbose", action="store_true")
    parser.add_argument("--log_dir", type=str, default="/content/logs_qnlp_pairs")
    args, _ = parser.parse_known_args()

    warnings.filterwarnings("ignore")
    os.environ["TOKENIZERS_PARALLELISM"] = "true"  # per lambeq tutorial recommendations

    np.random.seed(args.seed)

    df = read_csv_pairs(args.csv)

    # 80/20 split, stratified
    df_train, df_val = train_test_split(
        df,
        test_size=0.2,
        random_state=args.seed,
        stratify=df["label"],
    )

    train_s1 = df_train["sentence1"].tolist()
    train_s2 = df_train["sentence2"].tolist()
    train_y = df_train["label"].tolist()

    val_s1 = df_val["sentence1"].tolist()
    val_s2 = df_val["sentence2"].tolist()
    val_y = df_val["label"].tolist()

    # Tokenize + parse
    tokeniser = SpacyTokeniser()
    bobcat = BobcatParser(verbose="suppress")

    train_d1 = parse_sentences(train_s1, tokeniser, bobcat, verbose=args.verbose)
    train_d2 = parse_sentences(train_s2, tokeniser, bobcat, verbose=args.verbose)
    val_d1 = parse_sentences(val_s1, tokeniser, bobcat, verbose=args.verbose)
    val_d2 = parse_sentences(val_s2, tokeniser, bobcat, verbose=args.verbose)

    # Rewrite + ansatz
    rewriter = RemoveCupsRewriter()
    # Map NOUN and SENTENCE to 1 qubit each (common baseline for sentence-level circuits)
    mapping = {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1, AtomicType.PREPOSITIONAL_PHRASE:0}
    ansatz = IQPAnsatz(mapping, n_layers=1, n_single_qubit_params=3)

    pair_head = build_pair_head(n_params=4)

    train_circuits, train_labels = diagrams_to_pair_circuits(
        train_d1, train_d2, ansatz, rewriter, pair_head, train_y, verbose=args.verbose
    )
    val_circuits, val_labels = diagrams_to_pair_circuits(
        val_d1, val_d2, ansatz, rewriter, pair_head, val_y, verbose=args.verbose
    )

    if len(train_circuits) < 10 or len(val_circuits) < 5:
        raise RuntimeError(
            f"Too few parsable datapoints after filtering. "
            f"train={len(train_circuits)} val={len(val_circuits)}. "
            f"Consider cleaning text, verifying language, or loosening data constraints."
        )

    # Backend + model
    backend_config = make_backend_config(args.backend, args.shots)

    all_circuits = train_circuits + val_circuits
    model = TketModel.from_diagrams(all_circuits, backend_config=backend_config)

    # IMPORTANT: must initialize weights before forward/training
    model.initialise_weights()

    # Loss + metric (binary classification with 2D one-hot labels)
    bce = BinaryCrossEntropyLoss()

    def acc(y_hat: np.ndarray, y_true: np.ndarray) -> float:
        # y_hat and y_true are shape [batch, 2]
        # Using rounding + equality as in lambeq tutorial style
        return float(np.sum(np.round(y_hat) == y_true) / len(y_true) / 2)

    trainer = QuantumTrainer(
        model=model,
        loss_function=bce,
        epochs=args.epochs,
        optimizer=SPSAOptimizer,
        optim_hyperparams={
            "a": 0.05,
            "c": 0.06,
            "A": 0.001 * args.epochs,
        },
        evaluate_functions={"acc": acc},
        evaluate_on_train=True,
        verbose="text",
        log_dir=args.log_dir,
        seed=args.seed,
    )

    train_dataset = Dataset(train_circuits, train_labels, batch_size=args.batch_size)
    val_dataset = Dataset(val_circuits, val_labels, shuffle=False)

    # Training loop (trainer.fit)
    trainer.fit(
        train_dataset,
        val_dataset,
        early_stopping_criterion="acc",
        early_stopping_interval=5,
        minimize_criterion=False,
    )

    # Load best checkpoint if it exists
    best_path = os.path.join(trainer.log_dir, "best_model.lt")
    if os.path.exists(best_path):
        model.load(best_path)

    # Evaluate on validation set
    val_pred = model(val_circuits)  # shape [N, 2], probabilities
    y_true = np.argmax(np.array(val_labels), axis=1)
    y_pred = np.argmax(val_pred, axis=1)

    val_acc = accuracy_score(y_true, y_pred)
    print(f"\nValidation accuracy (argmax): {val_acc:.4f}")

    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    print("\nConfusion matrix (rows=true, cols=pred):")
    print(cm)

    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=CLASS_NAMES)
    disp.plot(values_format="d")
    plt.title("Confusion Matrix (Validation)")
    plt.tight_layout()
    plt.show()

    # Plot training curves
    plot_training_curves(trainer)


if __name__ == "__main__":
    main()
```

```python
# =============================
# 0) Install (Colab)
# =============================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from lambeq import (
    BobcatParser,
    RemoveCupsRewriter,
    AtomicType,
    IQPAnsatz,
    TketModel,
    QuantumTrainer,
    SPSAOptimizer,
    Dataset,
    BinaryCrossEntropyLoss,
)

from lambeq.backend import Symbol
from lambeq.backend.quantum import CX, Rz, Id, Discard, qubit

# =============================
# 1) Load CSV
# =============================
df = pd.read_csv('/content/DA_generarat.csv')

s1 = df['sentence1'].astype(str).tolist()
s2 = df['sentence2'].astype(str).tolist()
y  = df['label'].astype(int).tolist()   # 0/1

# one-hot labels pentru BCE (format ca în tutorial: [t, 1-t] sau [1,0]/[0,1])
# aici: 0 -> [1,0] (neutral), 1 -> [0,1] (implication)
Y = np.array([[1, 0] if lab == 0 else [0, 1] for lab in y], dtype=float)

# =============================
# 2) Stratified split 80/20
# =============================
s1_train, s1_val, s2_train, s2_val, Y_train, Y_val = train_test_split(
    s1, s2, Y, test_size=0.2, stratify=y, random_state=0
)

# =============================
# 3) Parse -> diagrams (cu filtrare None)
# =============================
parser = BobcatParser(verbose='suppress')

raw_train_d1 = parser.sentences2diagrams(s1_train, suppress_exceptions=True)
raw_train_d2 = parser.sentences2diagrams(s2_train, suppress_exceptions=True)
raw_val_d1   = parser.sentences2diagrams(s1_val,   suppress_exceptions=True)
raw_val_d2   = parser.sentences2diagrams(s2_val,   suppress_exceptions=True)

def filter_pairs(d1_list, d2_list, labels_2col):
    kept_d1, kept_d2, kept_y = [], [], []
    for d1, d2, lab in zip(d1_list, d2_list, labels_2col):
        if d1 is None or d2 is None:
            continue
        kept_d1.append(d1.normal_form())
        kept_d2.append(d2.normal_form())
        kept_y.append(lab)
    return kept_d1, kept_d2, np.array(kept_y, dtype=float)

train_d1, train_d2, Y_train_f = filter_pairs(raw_train_d1, raw_train_d2, Y_train)
val_d1,   val_d2,   Y_val_f   = filter_pairs(raw_val_d1,   raw_val_d2,   Y_val)

print(f"Train kept: {len(train_d1)} / {len(s1_train)}")
print(f"Val kept:   {len(val_d1)} / {len(s1_val)}")

# =============================
# 4) Remove cups (reduce post-selection)
# =============================
remove_cups = RemoveCupsRewriter()
train_d1 = [remove_cups(d) for d in train_d1]
train_d2 = [remove_cups(d) for d in train_d2]
val_d1   = [remove_cups(d) for d in val_d1]
val_d2   = [remove_cups(d) for d in val_d2]

# =============================
# 5) Ansatz + pair_head (2 circuite -> 1 qubit)
# =============================
# IMPORTANT:
# - În tutorialul quantum, ei pun SENTENCE: 0 când dataset-ul e NP/N (noun phrases). :contentReference[oaicite:1]{index=1}
# - La tine sunt propoziții, deci păstrăm SENTENCE: 1 (altfel riști să nu ai ieșire utilă).
mapping = {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1}

# Dacă ai AtomicType.PREPOSITIONAL_PHRASE în versiunea ta, îl setăm la 1 (mai safe decât 0).
if hasattr(AtomicType, "PREPOSITIONAL_PHRASE"):
    mapping[getattr(AtomicType, "PREPOSITIONAL_PHRASE")] = 1

ansatz = IQPAnsatz(mapping, n_layers=1, n_single_qubit_params=3)

# head cu 2 parametri antrenabili, apoi CX, apoi discard pe al doilea qubit => ieșire 1 qubit (2 clase)
syms = [Symbol('rel0'), Symbol('rel1')]
head = (Rz(syms[0]) @ Rz(syms[1])) >> CX >> (Id(qubit) @ Discard())

def make_pair_circuits(diags_a, diags_b):
    circs = []
    for d1, d2 in zip(diags_a, diags_b):
        c1 = ansatz(d1)
        c2 = ansatz(d2)
        combined = (c1 @ c2) >> head

        # Safety: dacă mai rămâne post_selection, îl golim ca să evităm crash-uri.
        # (Ideal: remove_cups + ansatz bun ar trebui să reducă mult post-selection-ul.)
        if hasattr(combined, "post_selection") and isinstance(combined.post_selection, dict) and combined.post_selection:
            combined.post_selection = {}

        circs.append(combined)
    return circs

train_circuits = make_pair_circuits(train_d1, train_d2)
val_circuits   = make_pair_circuits(val_d1, val_d2)

# =============================
# 6) Backend config + Model (TketModel)
# =============================
from pytket.extensions.qiskit import AerBackend

backend = AerBackend()
backend_config = {
    "backend": backend,
    "compilation": backend.default_compilation_pass(2),
    "shots": 2048,   # poți crește (4096/8192) pentru stabilitate
}

all_circuits = train_circuits + val_circuits
model = TketModel.from_diagrams(all_circuits, backend_config=backend_config)
model.initialise_weights()

# =============================
# 7) Datasets
# =============================
train_dataset = Dataset(train_circuits, Y_train_f, batch_size=16)
val_dataset   = Dataset(val_circuits,   Y_val_f,   shuffle=False)

# =============================
# 8) Trainer
# =============================
bce = BinaryCrossEntropyLoss()

# acc ca în tutorial (atenție: împărțire la 2 fiindcă se numără pe ambele coloane) :contentReference[oaicite:2]{index=2}
acc = lambda y_hat, y: np.sum(np.round(y_hat) == y) / len(y) / 2

trainer = QuantumTrainer(
    model,
    loss_function=bce,
    epochs=10,
    optimizer=SPSAOptimizer,
    optim_hyperparams={"a": 0.05, "c": 0.06, "A": 0.001 * 10},
    evaluate_functions={"acc": acc},
    evaluate_on_train=True,
    verbose="text",
    log_dir="logs_nli",
    seed=0
)

trainer.fit(
    train_dataset,
    val_dataset,
    early_stopping_criterion="acc",
    early_stopping_interval=5,
    minimize_criterion=False
)

# =============================
# 9) Eval rapid
# =============================
train_pred = model(train_circuits)
train_acc = acc(train_pred, Y_train_f)
print("Train acc:", float(train_acc))

val_pred = model(val_circuits)
val_acc = acc(val_pred, Y_val_f)
print("Val acc:", float(val_acc))
```

```python

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

from lambeq import (
    BobcatParser,
    RemoveCupsRewriter,
    AtomicType,
    IQPAnsatz,
    TketModel,
    QuantumTrainer,
    SPSAOptimizer,
    Dataset,
    BinaryCrossEntropyLoss,
)

from lambeq.backend import Symbol
from lambeq.backend.quantum import CX, Rz, Id, Discard, qubit

from pytket.extensions.qiskit import AerBackend


# =============================
# 1) Load CSV
# =============================
df = pd.read_csv('/content/DA_generarat.csv')
s1 = df['sentence1'].astype(str).tolist()
s2 = df['sentence2'].astype(str).tolist()
y  = df['label'].astype(int).tolist()   # 0/1

Y = np.array([[1, 0] if lab == 0 else [0, 1] for lab in y], dtype=float)

# =============================
# 2) Split stratificat
# =============================
s1_train, s1_val, s2_train, s2_val, y_train, y_val, Y_train, Y_val = train_test_split(
    s1, s2, y, Y, test_size=0.2, stratify=y, random_state=0
)

# =============================
# 3) Parse
# =============================
parser = BobcatParser(verbose='suppress')
d1_tr = parser.sentences2diagrams(s1_train, suppress_exceptions=True)
d2_tr = parser.sentences2diagrams(s2_train, suppress_exceptions=True)
d1_va = parser.sentences2diagrams(s1_val, suppress_exceptions=True)
d2_va = parser.sentences2diagrams(s2_val, suppress_exceptions=True)

def keep_valid(d1_list, d2_list, Y_list):
    D1, D2, YY = [], [], []
    for d1, d2, yy in zip(d1_list, d2_list, Y_list):
        if d1 is None or d2 is None:
            continue
        D1.append(d1.normal_form())
        D2.append(d2.normal_form())
        YY.append(yy)
    return D1, D2, np.array(YY, dtype=float)

d1_tr, d2_tr, Y_train = keep_valid(d1_tr, d2_tr, Y_train)
d1_va, d2_va, Y_val   = keep_valid(d1_va, d2_va, Y_val)

print(f"Train kept: {len(d1_tr)} / {len(s1_train)}")
print(f"Val kept:   {len(d1_va)} / {len(s1_val)}")

# =============================
# 4) Rewriting
# =============================
remove_cups = RemoveCupsRewriter()
d1_tr = [remove_cups(d) for d in d1_tr]
d2_tr = [remove_cups(d) for d in d2_tr]
d1_va = [remove_cups(d) for d in d1_va]
d2_va = [remove_cups(d) for d in d2_va]

# =============================
# 5) Ansatz + head
# =============================
mapping = {AtomicType.NOUN: 1, AtomicType.SENTENCE: 1}
if hasattr(AtomicType, "PREPOSITIONAL_PHRASE"):
    mapping[getattr(AtomicType, "PREPOSITIONAL_PHRASE")] = 1  # mai safe decât 0

ansatz = IQPAnsatz(mapping, n_layers=1, n_single_qubit_params=3)

syms = [Symbol('rel0'), Symbol('rel1')]
head = (Rz(syms[0]) @ Rz(syms[1])) >> CX >> (Id(qubit) @ Discard())  # output 1 qubit

def sanitize_post_selection(circ):
    """
    Taie intrările post_selection care cer biți inexistenți în bitstring.
    Asta previne IndexError la evaluare.
    """
    ps = getattr(circ, "post_selection", None)
    if not isinstance(ps, dict) or not ps:
        return circ

    # n_bits “măsurate” (în lambeq, asta e cel folosit când se iterează bitstring-ul)
    try:
        n_bits = len(circ.post_processing.dom)
    except Exception:
        # fallback: numărul de qubiți din codom
        try:
            n_bits = len(circ.cod)
        except Exception:
            circ.post_selection = {}
            return circ

    # păstrăm doar indecșii validați
    new_ps = {i: v for i, v in ps.items() if isinstance(i, int) and 0 <= i < n_bits}

    # dacă încă pare dubios, mai bine îl golești complet decât să crape
    circ.post_selection = new_ps
    return circ

def make_pair_circuits(D1, D2):
    out = []
    for d1, d2 in zip(D1, D2):
        c1 = ansatz(d1)
        c2 = ansatz(d2)
        combined = (c1 @ c2) >> head
        combined = sanitize_post_selection(combined)
        out.append(combined)
    return out

train_circuits = make_pair_circuits(d1_tr, d2_tr)
val_circuits   = make_pair_circuits(d1_va, d2_va)

# =============================
# 5.1) Debug: vezi dacă mai există post_selection invalid
# =============================
def debug_post_selection(circs, name):
    bad = 0
    for i, c in enumerate(circs):
        ps = getattr(c, "post_selection", {}) or {}
        if not ps:
            continue
        try:
            n_bits = len(c.post_processing.dom)
        except Exception:
            n_bits = None
        mx = max(ps.keys()) if ps else None
        if n_bits is not None and mx is not None and mx >= n_bits:
            bad += 1
            print(f"[{name}] BAD idx={i} n_bits={n_bits} post_selection={ps}")
    print(f"[{name}] circuits with (potentially) bad post_selection: {bad}")

debug_post_selection(train_circuits, "train")
debug_post_selection(val_circuits, "val")

# =============================
# 6) Backend + Model
# =============================
backend = AerBackend()
backend_config = {
    "backend": backend,
    "compilation": None,   # IMPORTANT: încearcă fără compilation (mai robust)
    "shots": 2048
}

all_circuits = train_circuits + val_circuits
model = TketModel.from_diagrams(all_circuits, backend_config=backend_config)
model.initialise_weights()

# =============================
# 7) Dataset + Trainer
# =============================
train_dataset = Dataset(train_circuits, Y_train, batch_size=16)
val_dataset   = Dataset(val_circuits,   Y_val, shuffle=False)

loss = BinaryCrossEntropyLoss()
acc = lambda y_hat, y: np.sum(np.round(y_hat) == y) / len(y) / 2

trainer = QuantumTrainer(
    model,
    loss_function=loss,
    epochs=10,
    optimizer=SPSAOptimizer,
    optim_hyperparams={"a": 0.05, "c": 0.06, "A": 0.01},
    evaluate_functions={"acc": acc},
    evaluate_on_train=True,
    verbose="text",
    seed=0
)

trainer.fit(
    train_dataset,
    val_dataset,
    early_stopping_criterion="acc",
    early_stopping_interval=5,
    minimize_criterion=False
)

# =============================
# 8) Eval
# =============================
train_pred = model(train_circuits)
print("Train acc:", float(acc(train_pred, Y_train)))

val_pred = model(val_circuits)
print("Val acc:", float(acc(val_pred, Y_val)))
```

```python
# ============================================================
# LAMBEQ QUANTUM TRAINER – NLI din CSV (SINGLE CELL, STABIL)
# Fix-uri principale:
#  - output pe 2 clase: AtomicType.SENTENCE = 2 (match cu one-hot [N,2])
#  - acc corect (argmax)
#  - protecție dacă rămân prea puține diagrame după parse/filter
# ============================================================

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

from lambeq import (
    BobcatParser,
    RemoveCupsRewriter,
    AtomicType,
    IQPAnsatz,
    TketModel,
    QuantumTrainer,
    SPSAOptimizer,
    Dataset,
    BinaryCrossEntropyLoss,
)

from pytket.extensions.qiskit import AerBackend


# ------------------------------------------------------------
# 1) Load CSV + concat propoziții (workaround stabil lambeq)
# ------------------------------------------------------------
df = pd.read_csv('/content/DA_generarat.csv')

# asigură existența coloanelor
required_cols = {"sentence1", "sentence2", "label"}
missing = required_cols - set(df.columns)
if missing:
    raise ValueError(f"Lipsesc coloane în CSV: {missing}. Ai coloanele: {list(df.columns)}")

pairs = [
    f"{a} implies {b}"
    for a, b in zip(df.sentence1.astype(str), df.sentence2.astype(str))
]

labels = df.label.astype(int).values

# one-hot pe 2 clase (0/1)
Y = np.array([[1, 0] if y == 0 else [0, 1] for y in labels], dtype=np.float32)


# ------------------------------------------------------------
# 2) Train / Val split
# ------------------------------------------------------------
pairs_train, pairs_val, Y_train, Y_val, labels_train, labels_val = train_test_split(
    pairs, Y, labels, test_size=0.2, stratify=labels, random_state=0
)


# ------------------------------------------------------------
# 3) Parse sentences → DisCoCat diagrams
# ------------------------------------------------------------
parser = BobcatParser(verbose='suppress')

train_diagrams = parser.sentences2diagrams(pairs_train, suppress_exceptions=True)
val_diagrams   = parser.sentences2diagrams(pairs_val,   suppress_exceptions=True)

def filter_none(diagrams, labels_onehot):
    d, l = [], []
    for x, y in zip(diagrams, labels_onehot):
        if x is not None:
            d.append(x.normal_form())
            l.append(y)
    return d, np.array(l, dtype=np.float32)

train_diagrams, Y_train = filter_none(train_diagrams, Y_train)
val_diagrams,   Y_val   = filter_none(val_diagrams,   Y_val)

print("Train diagrams:", len(train_diagrams))
print("Val diagrams:",   len(val_diagrams))

if len(train_diagrams) < 2 or len(val_diagrams) < 2:
    raise RuntimeError(
        "Au rămas prea puține diagrame după parsing/filtering. "
        "Încearcă să cureți textul, să reduci lungimea propozițiilor, "
        "sau să schimbi formatul (ex: 'premise: ... hypothesis: ...')."
    )


# ------------------------------------------------------------
# 4) Remove cups (reduce post-selection)
# ------------------------------------------------------------
rewriter = RemoveCupsRewriter()
train_diagrams = [rewriter(d) for d in train_diagrams]
val_diagrams   = [rewriter(d) for d in val_diagrams]


# ------------------------------------------------------------
# 5) Ansatz → Quantum circuits
# IMPORTANT: SENTENCE=2 ca să corespundă cu Y one-hot [N,2]
# ------------------------------------------------------------
mapping = {AtomicType.NOUN: 1, AtomicType.SENTENCE: 2}
ansatz = IQPAnsatz(mapping, n_layers=1, n_single_qubit_params=3)

train_circuits = [ansatz(d) for d in train_diagrams]
val_circuits   = [ansatz(d) for d in val_diagrams]


# ------------------------------------------------------------
# 6) Backend + Quantum model
# ------------------------------------------------------------
backend = AerBackend()

backend_config = {
    "backend": backend,
    "compilation": None,   # previne crash-uri lambeq/tket în unele setup-uri
    "shots": 2048
}

# în multe versiuni lambeq, lista asta e acceptată; dacă ai TypeError aici,
# spune-mi exact traceback-ul și îți adaptez pe API-ul tău instalat.
model = TketModel.from_diagrams(
    train_circuits + val_circuits,
    backend_config=backend_config
)
model.initialise_weights()


# ------------------------------------------------------------
# 7) Dataset + Trainer
# ------------------------------------------------------------
train_dataset = Dataset(train_circuits, Y_train, batch_size=16)
val_dataset   = Dataset(val_circuits,   Y_val, shuffle=False)

bce = BinaryCrossEntropyLoss()

def acc(y_hat, y):
    # y_hat și y sunt [N,2]
    return float((np.argmax(y_hat, axis=1) == np.argmax(y, axis=1)).mean())

trainer = QuantumTrainer(
    model,
    loss_function=bce,
    epochs=15,
    optimizer=SPSAOptimizer,
    optim_hyperparams={'a': 0.05, 'c': 0.06, 'A': 0.01},
    evaluate_functions={'acc': acc},
    evaluate_on_train=True,
    verbose='text',
    seed=0
)

trainer.fit(train_dataset, val_dataset)


# ------------------------------------------------------------
# 8) Final evaluation
# ------------------------------------------------------------
train_pred = model(train_circuits)
val_pred   = model(val_circuits)

print("\nFinal train acc:", acc(train_pred, Y_train))
print("Final val acc:",   acc(val_pred,   Y_val))
```

```python
# ============================================================
# LAMBEQ QUANTUM TRAINER – NLI din CSV
# NU mai combin propozițiile în text.
# Parsez separat sentence1 și sentence2, apoi "combin" diagramele/circuitele
# prin tensor product ( @ ) după ansatz.
# ============================================================

import warnings
warnings.filterwarnings("ignore", category=FutureWarning)

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

from lambeq import (
    BobcatParser,
    RemoveCupsRewriter,
    AtomicType,
    IQPAnsatz,
    TketModel,
    QuantumTrainer,
    SPSAOptimizer,
    Dataset,
    BinaryCrossEntropyLoss,
)

from pytket.extensions.qiskit import AerBackend


# ---------------------------
# Utils
# ---------------------------
def acc(y_hat, y):
    return float((np.argmax(y_hat, axis=1) == np.argmax(y, axis=1)).mean())

def safe_remove_cups_pair(d1, d2):
    """RemoveCupsRewriter pe perechi; dacă crapă, întoarce None."""
    rewriter = RemoveCupsRewriter()
    try:
        return rewriter(d1).normal_form(), rewriter(d2).normal_form()
    except Exception:
        return None

def filter_large_circuits(circuits, labels_onehot, max_qubits=26):
    keep_c, keep_y = [], []
    dropped = 0
    max_seen = 0
    for c, y in zip(circuits, labels_onehot):
        nq = len(c.qubits)
        max_seen = max(max_seen, nq)
        if nq <= max_qubits:
            keep_c.append(c)
            keep_y.append(y)
        else:
            dropped += 1
    print(f"Max qubits seen: {max_seen}")
    print(f"Dropped {dropped}/{len(circuits)} circuits with > {max_qubits} qubits")
    return keep_c, np.array(keep_y, dtype=np.float32)


# ------------------------------------------------------------
# 1) Load CSV (FĂRĂ concatenare)
# ------------------------------------------------------------
df = pd.read_csv('/content/DA_generarat.csv')

required_cols = {"sentence1", "sentence2", "label"}
missing = required_cols - set(df.columns)
if missing:
    raise ValueError(f"Lipsesc coloane în CSV: {missing}. Ai: {list(df.columns)}")

s1 = df.sentence1.astype(str).tolist()
s2 = df.sentence2.astype(str).tolist()

labels = df.label.astype(int).values
Y = np.array([[1, 0] if y == 0 else [0, 1] for y in labels], dtype=np.float32)


# ------------------------------------------------------------
# 2) Split pe indecși (ca să păstrăm perechile)
# ------------------------------------------------------------
idx = np.arange(len(df))
idx_train, idx_val = train_test_split(
    idx, test_size=0.2, stratify=labels, random_state=0
)

s1_train = [s1[i] for i in idx_train]
s2_train = [s2[i] for i in idx_train]
Y_train  = Y[idx_train]

s1_val = [s1[i] for i in idx_val]
s2_val = [s2[i] for i in idx_val]
Y_val  = Y[idx_val]


# ------------------------------------------------------------
# 3) Parse separat sentence1 și sentence2
# ------------------------------------------------------------
parser = BobcatParser(verbose='suppress')

d1_train = parser.sentences2diagrams(s1_train, suppress_exceptions=True)
d2_train = parser.sentences2diagrams(s2_train, suppress_exceptions=True)
d1_val   = parser.sentences2diagrams(s1_val,   suppress_exceptions=True)
d2_val   = parser.sentences2diagrams(s2_val,   suppress_exceptions=True)

# filtrăm perechi unde oricare e None
pairs_train, Y_train_f = [], []
for a, b, y in zip(d1_train, d2_train, Y_train):
    if a is not None and b is not None:
        pairs_train.append((a.normal_form(), b.normal_form()))
        Y_train_f.append(y)
pairs_val, Y_val_f = [], []
for a, b, y in zip(d1_val, d2_val, Y_val):
    if a is not None and b is not None:
        pairs_val.append((a.normal_form(), b.normal_form()))
        Y_val_f.append(y)

Y_train = np.array(Y_train_f, dtype=np.float32)
Y_val   = np.array(Y_val_f, dtype=np.float32)

print("Train pairs after parse:", len(pairs_train))
print("Val pairs after parse:",   len(pairs_val))

if len(pairs_train) < 5 or len(pairs_val) < 5:
    raise RuntimeError("Prea puține perechi după parsing. Datele nu se parsează ok.")


# ------------------------------------------------------------
# 4) Remove cups – SAFE pe perechi (dacă scoate prea multe, îl sărim)
# ------------------------------------------------------------
pairs_train_rc, Y_train_rc = [], []
dropped = 0
for (a, b), y in zip(pairs_train, Y_train):
    out = safe_remove_cups_pair(a, b)
    if out is None:
        dropped += 1
        continue
    pairs_train_rc.append(out)
    Y_train_rc.append(y)
pairs_val_rc, Y_val_rc = [], []
dropped2 = 0
for (a, b), y in zip(pairs_val, Y_val):
    out = safe_remove_cups_pair(a, b)
    if out is None:
        dropped2 += 1
        continue
    pairs_val_rc.append(out)
    Y_val_rc.append(y)

print(f"RemoveCupsRewriter dropped train: {dropped}/{len(pairs_train)}")
print(f"RemoveCupsRewriter dropped val:   {dropped2}/{len(pairs_val)}")

# folosim output-ul rewriter-ului doar dacă rămâne suficient
if len(pairs_train_rc) >= 5 and len(pairs_val_rc) >= 5:
    pairs_train = pairs_train_rc
    pairs_val   = pairs_val_rc
    Y_train = np.array(Y_train_rc, dtype=np.float32)
    Y_val   = np.array(Y_val_rc, dtype=np.float32)
    print("Using RemoveCupsRewriter output ✅")
else:
    print("Skipping RemoveCupsRewriter (kept original diagrams) ✅")


# ------------------------------------------------------------
# 5) Ansatz → circuite pe fiecare propoziție, apoi combinăm prin tensor (@)
# ------------------------------------------------------------
mapping = {AtomicType.NOUN: 1, AtomicType.SENTENCE: 2, AtomicType.PREPOSITIONAL_PHRASE:0}
ansatz = IQPAnsatz(mapping, n_layers=1, n_single_qubit_params=3)

train_circuits = []
for d1, d2 in pairs_train:
    c1 = ansatz(d1)
    c2 = ansatz(d2)
    train_circuits.append(c1 @ c2)   # <-- combinare "după" parse, nu în text

val_circuits = []
for d1, d2 in pairs_val:
    c1 = ansatz(d1)
    c2 = ansatz(d2)
    val_circuits.append(c1 @ c2)

# opțional dar recomandat: taie circuite prea mari ca să eviți OOM
train_circuits, Y_train = filter_large_circuits(train_circuits, Y_train, max_qubits=26)
val_circuits,   Y_val   = filter_large_circuits(val_circuits,   Y_val,   max_qubits=26)

if len(train_circuits) < 5 or len(val_circuits) < 5:
    raise RuntimeError("Au rămas prea puține circuite după filtrarea de qubiți.")


# ------------------------------------------------------------
# 6) Backend + Model (shots => sampling, nu statevector)
# ------------------------------------------------------------
backend = AerBackend()
backend_config = {
    "backend": backend,
    "compilation": None,
    "shots": 2048
}

model = TketModel.from_diagrams(
    train_circuits + val_circuits,
    backend_config=backend_config
)
model.initialise_weights()


# ------------------------------------------------------------
# 7) Dataset + Trainer
# ------------------------------------------------------------
train_dataset = Dataset(train_circuits, Y_train, batch_size=16)
val_dataset   = Dataset(val_circuits,   Y_val, shuffle=False)

trainer = QuantumTrainer(
    model,
    loss_function=BinaryCrossEntropyLoss(),
    epochs=10,
    optimizer=SPSAOptimizer,
    optim_hyperparams={'a': 0.05, 'c': 0.06, 'A': 0.01},
    evaluate_functions={'acc': acc},
    evaluate_on_train=True,
    verbose='text',
    seed=0
)

trainer.fit(train_dataset, val_dataset)


# ------------------------------------------------------------
# 8) Final evaluation
# ------------------------------------------------------------
train_pred = model(train_circuits)
val_pred   = model(val_circuits)

print("\nFinal train acc:", acc(train_pred, Y_train))
print("Final val acc:",   acc(val_pred,   Y_val))
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

import os
import json
import argparse
import random
from dataclasses import asdict, dataclass
from pathlib import Path
from typing import List, Tuple, Optional

import numpy as np
import pandas as pd
import torch
from torch import nn

from tqdm import tqdm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    classification_report
)

from lambeq import (
    BobcatParser,
    SpacyTokeniser,
    RemoveCupsRewriter,
    IQPAnsatz,
    AtomicType,
    UnknownWordsRewriteRule,
    Rewriter,
    Dataset,
    PennyLaneModel,
    PytorchTrainer,
)
from lambeq.backend.numerical_backend import set_backend as lambeq_set_backend


# ---------------- CONFIG ----------------

@dataclass
class Config:
    csv_path: str
    out_dir: str = "./outputs/run_qnlp_5class"
    seed: int = 42
    val_size: float = 0.2

    batch_size: int = 8
    epochs: int = 40
    learning_rate: float = 0.02

    min_freq_unk: int = 2


# ---------------- SEED ----------------

def set_seed(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


# ---------------- DATA ----------------

def clean(x):
    if not isinstance(x, str):
        return None
    return " ".join(x.strip().split())


def load_csv(path):
    df = pd.read_csv(path)

    required = {"sentence1", "sentence2", "label"}
    if not required.issubset(df.columns):
        raise ValueError("CSV must have sentence1, sentence2, label")

    df["sentence1"] = df["sentence1"].apply(clean)
    df["sentence2"] = df["sentence2"].apply(clean)

    df = df.dropna().reset_index(drop=True)

    df["label"] = df["label"].astype(int)

    if not set(df["label"].unique()).issubset({0, 1, 2, 3, 4}):
        raise ValueError("Labels must be in range 0–4")

    return df.sample(frac=1.0, random_state=42).reset_index(drop=True)


# ---------------- MODEL ----------------

class PairModel5Class(PennyLaneModel):
    """
    Pair model -> 5-class classification
    """

    def __init__(self, hidden_dim=16, dropout=0.1, **kwargs):
        super().__init__(**kwargs)

        self.net = nn.Sequential(
            nn.Linear(4, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 5)   # 🔥 5 CLASE
        )

    def forward(self, diagram_pairs):
        d1, d2 = zip(*diagram_pairs)

        o1 = self.get_diagram_output(list(d1))  # [B,2]
        o2 = self.get_diagram_output(list(d2))  # [B,2]

        x = torch.cat([o1, o2], dim=1)          # [B,4]

        logits = self.net(x)                    # [B,5]
        return logits


# ---------------- EVAL ----------------

def evaluate_multiclass(y_true, logits):
    y_pred = np.argmax(logits, axis=1)

    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, average="macro", zero_division=0
    )

    return {
        "accuracy": float(acc),
        "precision_macro": float(prec),
        "recall_macro": float(rec),
        "f1_macro": float(f1),
        "confusion_matrix": confusion_matrix(y_true, y_pred).tolist(),
        "classification_report": classification_report(y_true, y_pred, digits=4),
    }


# ---------------- MAIN ----------------

def main(cfg: Config):

    set_seed(cfg.seed)
    out_dir = Path(cfg.out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    df = load_csv(cfg.csv_path)

    train_df, val_df = train_test_split(
        df,
        test_size=cfg.val_size,
        stratify=df["label"],
        random_state=cfg.seed
    )

    # ---------------- PARSER ----------------
    tokeniser = SpacyTokeniser()
    parser = BobcatParser()

    def parse(sentences):
        out = []
        for s in tqdm(sentences):
            try:
                toks = tokeniser.tokenise_sentence(s)
                out.append(parser.sentence2diagram(toks, tokenised=True))
            except:
                out.append(None)
        return out

    train_df["d1"] = parse(train_df["sentence1"])
    train_df["d2"] = parse(train_df["sentence2"])
    val_df["d1"] = parse(val_df["sentence1"])
    val_df["d2"] = parse(val_df["sentence2"])

    train_df = train_df.dropna()
    val_df = val_df.dropna()

    # ---------------- UNK ----------------
    unk = UnknownWordsRewriteRule.from_diagrams(
        train_df["d1"].tolist() + train_df["d2"].tolist(),
        min_freq=cfg.min_freq_unk
    )
    rewriter = Rewriter([unk])

    remove_cups = RemoveCupsRewriter()

    ansatz = IQPAnsatz(
        {
            AtomicType.NOUN: 1,
            AtomicType.SENTENCE: 1,
            AtomicType.PREPOSITIONAL_PHRASE: 0
        },
        n_layers=1
    )

    def to_circuit(diag):
        d = rewriter(diag)
        d = remove_cups(d)
        return ansatz(d)

    train_df["c1"] = train_df["d1"].apply(to_circuit)
    train_df["c2"] = train_df["d2"].apply(to_circuit)
    val_df["c1"] = val_df["d1"].apply(to_circuit)
    val_df["c2"] = val_df["d2"].apply(to_circuit)

    # ---------------- DATASET ----------------
    lambeq_set_backend("pytorch")

    train_pairs = list(zip(train_df["c1"], train_df["c2"]))
    val_pairs = list(zip(val_df["c1"], val_df["c2"]))

    y_train = train_df["label"].tolist()
    y_val = val_df["label"].tolist()

    all_circuits = train_df["c1"].tolist() + train_df["c2"].tolist() + \
                   val_df["c1"].tolist() + val_df["c2"].tolist()

    model = PairModel5Class.from_diagrams(
        all_circuits,
        probabilities=True,
        backend_config={"backend": "default.qubit"},
        hidden_dim=16,
        dropout=0.1
    )

    model.initialise_weights()

    train_dataset = Dataset(train_pairs, y_train, batch_size=cfg.batch_size, shuffle=True)
    val_dataset = Dataset(val_pairs, y_val, batch_size=cfg.batch_size)

    # ---------------- LOSS (IMPORTANT) ----------------
    loss_fn = torch.nn.CrossEntropyLoss()

    def acc_fn(logits, y):
        preds = torch.argmax(logits, dim=1)
        return (preds == y).float().mean().item()

    trainer = PytorchTrainer(
        model=model,
        loss_function=loss_fn,
        optimizer=torch.optim.Adam,
        learning_rate=cfg.learning_rate,
        epochs=cfg.epochs,
        evaluate_functions={"acc": acc_fn},
        evaluate_on_train=True,
        verbose="text"
    )

    trainer.fit(train_dataset, val_dataset)

    # ---------------- EVAL ----------------
    model.eval()
    with torch.no_grad():
        logits = model(val_pairs).cpu().numpy()

    metrics = evaluate_multiclass(np.array(y_val), logits)

    print("\n=== RESULTS ===")
    print(metrics["classification_report"])

    with open(out_dir / "metrics.json", "w") as f:
        json.dump(metrics, f, indent=2)

    torch.save(model.state_dict(), out_dir / "model.pt")


# ---------------- RUN ----------------

if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--csv_path", type=str, default="/content/AllClasses.csv")
    parser.add_argument("--out_dir", type=str, default="./out")

    args, _ = parser.parse_known_args()

    cfg = Config(**vars(args))
    main(cfg)
```
